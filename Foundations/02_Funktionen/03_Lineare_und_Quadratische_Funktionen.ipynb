## Teil A: Lineare Funktionen

### 1. Definition und Parameter

Eine **lineare Funktion** hat die allgemeine Form:

$$f(x) = kx + d$$

| Parameter | Bedeutung | Grafisch |
|-----------|-----------|----------|
| $k$ | **Steigung** (Änderungsrate) | Neigung der Geraden |
| $d$ | **y-Achsenabschnitt** | Schnittpunkt mit der y-Achse |

**Steigung $k$:**  
Die Steigung gibt an, um wie viele Einheiten sich $f(x)$ ändert, wenn $x$ um 1 zunimmt.

$$k = \frac{\Delta y}{\Delta x} = \frac{f(x_2) - f(x_1)}{x_2 - x_1}$$

Ist $k > 0$, steigt die Gerade von links nach rechts; ist $k < 0$, fällt sie. Für $k = 0$ ergibt sich eine horizontale Gerade. Je größer $|k|$, desto steiler der Graph.

**Nullstelle:** $f(x) = 0 \Rightarrow x = -\frac{d}{k}$ (für $k \ne 0$)

---

### Geradengleichung aufstellen

Kennt man die Steigung und einen Punkt - oder zwei Punkte - lässt sich die Geradengleichung direkt aufstellen:

| Gegeben | Methode |
|---------|---------|
| Steigung $k$ + Punkt $P(x_0, y_0)$ | $f(x) = k(x - x_0) + y_0$ |
| Zwei Punkte $P_1, P_2$ | $k = \frac{y_2 - y_1}{x_2 - x_1}$, dann einsetzen |

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Lineare Funktionen: Einfluss von k und d ---

def visualisiere_lineare_funktionen():
    """Zeigt den Einfluss der Parameter k und d auf die Gerade."""
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    x = np.linspace(-4, 4, 300)

    # --- Links: Verschiedene Steigungen k (d=0) ---
    ax = axes[0]
    steigungen = [-2, -0.5, 0, 0.5, 1, 2]
    farben = plt.cm.coolwarm(np.linspace(0, 1, len(steigungen)))
    for k, farbe in zip(steigungen, farben):
        ax.plot(x, k*x, color=farbe, linewidth=2, label=f"k = {k}")
    ax.set_title("Einfluss der Steigung k\n" + r"$f(x) = kx$", fontsize=13)
    ax.legend(fontsize=10, loc='lower right')
    ax.set_ylim(-6, 6)

    # --- Rechts: Verschiedene d (k=1) ---
    ax = axes[1]
    d_werte = [-3, -1, 0, 1, 3]
    farben = plt.cm.viridis(np.linspace(0, 1, len(d_werte)))
    for d, farbe in zip(d_werte, farben):
        ax.plot(x, x + d, color=farbe, linewidth=2, label=f"d = {d}")
        ax.scatter(0, d, color=farbe, s=80, zorder=5)
    ax.set_title("Einfluss des y-Achsenabschnitts d\n" + r"$f(x) = x + d$", fontsize=13)
    ax.legend(fontsize=10)
    ax.set_ylim(-6, 6)

    for ax in axes:
        ax.grid(True, alpha=0.3)
        ax.axhline(0, color='black', linewidth=0.5)
        ax.axvline(0, color='black', linewidth=0.5)
        ax.set_xlabel("x")
        ax.set_ylabel("f(x)")

    plt.tight_layout()
    plt.show()

visualisiere_lineare_funktionen()
```

## Teil B: Quadratische Funktionen

### 2. Die drei Darstellungsformen

| Form | Formel | Daraus direkt ablesbar |
|------|--------|------------------------|
| **Normalform** | $f(x) = ax^2 + bx + c$ | y-Achsenabschnitt $c$ |
| **Scheitelpunktform** | $f(x) = a(x - h)^2 + k$ | Scheitelpunkt $S(h, k)$ |
| **Produktform** | $f(x) = a(x - x_1)(x - x_2)$ | Nullstellen $x_1, x_2$ |

**Parameter $a$:**

| $a$ | Wirkung |
|-----|--------|
| $a > 0$ | Parabel nach oben geöffnet (U-Form) |
| $a < 0$ | Parabel nach unten geöffnet (∩-Form) |
| $\|a\| > 1$ | Schmaler (gestreckt) |
| $\|a\| < 1$ | Breiter (gestaucht) |

---

### 3. Scheitelpunkt

Der **Scheitelpunkt** $S(h, k)$ ist der höchste oder tiefste Punkt der Parabel.

Aus der Normalform $ax^2 + bx + c$:

$$h = -\frac{b}{2a}, \qquad k = f(h) = c - \frac{b^2}{4a}$$

---

### 4. Nullstellen und Diskriminante

Wie viele Schnittpunkte hat eine Parabel mit der x-Achse? Die Antwort steckt vollständig in der **Diskriminante**.

Die Nullstellen von $ax^2 + bx + c = 0$ ergeben sich aus der **Lösungsformel**:

> Siehe Algebra/Quadratische Gleichungen für Details.

$$x_{1,2} = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}$$

Die **Diskriminante** $D = b^2 - 4ac$ bestimmt die Anzahl der Lösungen:

| $D$ | Nullstellen | Grafisch |
|-----|-------------|----------|
| $D > 0$ | Zwei verschiedene | Parabel schneidet x-Achse zweimal |

| $D = 0$ | Eine doppelte | Parabel berührt x-Achse || $D < 0$ | Keine reelle | Parabel schneidet x-Achse nicht |

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Quadratische Funktionen: Formen und Diskriminante ---

def visualisiere_parabeln():
    """Zeigt verschiedene Parabeln und die drei Diskriminante-Fälle."""
    fig, axes = plt.subplots(1, 3, figsize=(19.5, 6))
    x = np.linspace(-4, 6, 500)

    # --- D > 0: Zwei Nullstellen ---
    ax = axes[0]
    y = x**2 - 4*x + 3  # Nullstellen bei x=1, x=3
    ax.plot(x, y, 'b-', linewidth=2.5)
    ax.scatter([1, 3], [0, 0], color='red', s=100, zorder=5, label="Nullstellen")
    ax.scatter([2], [-1], color='green', s=120, zorder=5, marker='s', label="Scheitelpunkt")
    ax.annotate("S(2, −1)", xy=(2, -1), xytext=(2.5, -2.5), fontsize=11,
                arrowprops=dict(arrowstyle='->', color='green'))
    ax.set_title(r"$D > 0$: Zwei Nullstellen" + "\n" + r"$x^2 - 4x + 3$", fontsize=13)
    ax.legend(fontsize=10)
    ax.set_ylim(-3, 8)

    # --- D = 0: Eine doppelte Nullstelle ---
    ax = axes[1]
    y = x**2 - 4*x + 4  # Nullstelle bei x=2 (doppelt)
    ax.plot(x, y, 'b-', linewidth=2.5)
    ax.scatter([2], [0], color='red', s=100, zorder=5, label="Doppelte Nullstelle")
    ax.set_title(r"$D = 0$: Berührt x-Achse" + "\n" + r"$x^2 - 4x + 4 = (x-2)^2$", fontsize=13)
    ax.legend(fontsize=10)
    ax.set_ylim(-1, 8)

    # --- D < 0: Keine reelle Nullstelle ---
    ax = axes[2]
    y = x**2 - 4*x + 5  # D = 16-20 = -4
    ax.plot(x, y, 'b-', linewidth=2.5)
    ax.scatter([2], [1], color='green', s=120, zorder=5, marker='s', label="Scheitel über x-Achse")
    ax.set_title(r"$D < 0$: Keine reelle Nullstelle" + "\n" + r"$x^2 - 4x + 5$", fontsize=13)
    ax.legend(fontsize=10)
    ax.set_ylim(-1, 8)

    for ax in axes:
        ax.grid(True, alpha=0.3)
        ax.axhline(0, color='black', linewidth=0.8)
        ax.axvline(0, color='black', linewidth=0.5)
        ax.set_xlabel("x")
        ax.set_ylabel("f(x)")

    plt.tight_layout()
    plt.show()

visualisiere_parabeln()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Einfluss des Parameters a ---

def visualisiere_parameter_a():
    """Zeigt den Einfluss von a auf Öffnung und Breite der Parabel."""
    fig, ax = plt.subplots(figsize=(6, 4))
    x = np.linspace(-3, 3, 300)

    a_werte = [-2, -0.5, 0.25, 0.5, 1, 2]
    farben = plt.cm.coolwarm(np.linspace(0, 1, len(a_werte)))

    for a, farbe in zip(a_werte, farben):
        ax.plot(x, a*x**2, color=farbe, linewidth=2, label=f"a = {a}")

    ax.set_title(r"Einfluss von $a$ auf die Parabel $f(x) = ax^2$", fontsize=14)
    ax.legend(fontsize=10, loc='lower right', ncol=2)
    ax.grid(True, alpha=0.3)
    ax.axhline(0, color='black', linewidth=0.8)
    ax.axvline(0, color='black', linewidth=0.5)
    ax.set_xlabel("x")
    ax.set_ylabel("f(x)")
    ax.set_ylim(-6, 6)

    plt.tight_layout()
    plt.show()

visualisiere_parameter_a()
```

## Anwendungen Quadratischer Funktionen

### 1. Physik: Projektilbewegung
Die Höhe eines geworfenen Objekts folgt einer quadratischen Funktion:
$$h(t) = h_0 + v_0 \,t - \frac{1}{2}g \,t^2$$

wobei:
- $h_0$ = Anfangshöhe
- $v_0$ = Anfangsgeschwindigkeit
- $g$ = Gravitationsbeschleunigung (=~ 9.81 m/s²)

Die Nullstellen zeigen, wann das Objekt den Boden berührt.

### 2. Geometrie: Flächenoptimierung
Bei rechteckigen Grundstücken mit begrenztem Umfang ist die Fläche eine quadratische Funktion:
$$A(x) = x(U/2 - x)$$

Der Scheitelpunkt zeigt die Abmessungen für maximale Fläche.

### 3. Optik & Akustik: Parabolische Reflektoren
Die Form von Satellitenschüsseln, Spiegeln und Mikrofonen folgt Parabeln. Der Grund ist geometrisch elegant: Alle von einer Parabel reflektierten parallelen Strahlen treffen sich exakt in ihrem **Brennpunkt** - das macht Parabeln ideal zum Bündeln von Licht, Schall oder Funksignalen. Die Gleichung $f(x) = ax^2$ bestimmt dabei, wie spitz oder flach die Parabel und damit der Fokus ist.

```python
import numpy as np
import matplotlib.pyplot as plt

def solve_projectile_problem():
    """
    Ein Ball wird von einer Brücke (5m hoch) mit 10m/s geworfen.
    Wann kommt der Ball auf dem Boden an?
    """

    print("PROBLEM: Projektilbwewegung")
    print("-" * 50)

    h0 = 5    # Anfangshöhe in Metern
    v0 = 10   # Anfangsgeschwindigkeit in m/s
    g = 9.81  # Gravitation

    # h(t) = h0 + v0*t - 0.5*g*t²
    # h(t) = 5 + 10*t - 4.905*t²

    a = -g/2
    b = v0
    c = h0

    print(f"h(t) = {c} + {b}t + {a}t²")
    print(f"h(t) = {a}t² + {b}t + {c}")

    # Nullstellen finden
    discriminant = b**2 - 4*a*c
    t1 = (-b - np.sqrt(discriminant)) / (2*a)
    t2 = (-b + np.sqrt(discriminant)) / (2*a)

    print(f"\nDiskriminante: Δ = {discriminant:.2f}")
    print(f"Nullstellen: t1 = {t1:.2f}, t2 = {t2:.2f}s")
    print(f"\n Der Ball kommt nach {max(t1, t2):.2f} Sekunden auf dem Boden an.")

    t = np.linspace(0, max(t1, t2) + 0.5, 100)
    h = c + b*t + a*t**2

    plt.figure(figsize=(6,4))
    plt.plot(t, h, 'b-', linewidth=2, label='h(t) = 5 + 10t - 4.905t²')
    plt.axhline(0, color='k', linewidth=0.5)
    plt.plot(max(t1, t2), 0, 'ro', markersize=10, label=f'Aufprall: t = {max(t1, t2):.2f}s')

    # Maximale Höhe (Scheitelpunkt)
    t_max = -b / (2*a)
    h_max = c + b*t_max + a*t_max**2
    plt.plot(t_max, h_max, 'go', markersize=10, label=f'Max. Höhe: h = {h_max:.2f}m bei t = {t_max:.2f}s')

    plt.xlabel('Zeit (s)')
    plt.ylabel('Höhe (m)')
    plt.title('Projektilbewegung: Ball von Brücke')
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

def solve_optimization_problem():
    """
    Ein Bauer hat 40m Zaun für ein reckteckiges Gehege.
    Wie sind die optimalen Abmessungen?
    """
    print("\n\nPROBLEM: Flächenoptimierung")
    print("-" * 50)

    perimeter = 40 # Gesamtzaun

    # Wenn Länge = x, dann Breite = (P/2 - x) = (20 - x)
    # Fläche A(x) = x * (20 - x) = -x² + 20x

    a = -1
    b = 20
    c = 0

    print(f"Umnfang = {perimeter}m")
    print(f"A(x) = x(20 . x ) = {a}x² + {b}x + {c}")

    # Scheitelpunkt (Maximum)
    x_optimal = -b / (2*a)
    area_max = c + b*x_optimal + a*x_optimal**2

    print(f"\nOptimale Länge: x = {x_optimal}m")
    print(f"Optimale Breite: {20 - x_optimal}m")
    print(f"Maximale Fläche: A = {area_max}m²")

    # Visualisierung
    x = np.linspace(0, 20, 100)
    area = -x**2 + 20*x

    plt.figure(figsize=(6,4))
    plt.plot(x, area, 'b-', linewidth=2, label='A(x) = -x² + 20x')
    plt.plot(x_optimal, area_max, 'ro', markersize=10, label=f'Maximum: A = {area_max}m² bei x = {x_optimal}m')
    plt.axvline(x_optimal, color='r', linestyle='--', alpha=0.5)

    plt.xlabel('Länge (m)')
    plt.ylabel('Fläche (m²)')
    plt.title('Flächenoptimierung: Rechteckiges Gehege mit 40m Zaun')
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

solve_projectile_problem()
solve_optimization_problem()
```

## Zusammenfassung

| Thema | Kernformel |
|-------|------------|
| **Lineare Funktion** | $f(x) = kx + d$ - Steigung $k$, Abschnitt $d$ |
| **Geradenbestimmung** | $k = \frac{\Delta y}{\Delta x}$ aus zwei Punkten |
| **Normalform** | $f(x) = ax^2 + bx + c$ |
| **Scheitelpunktform** | $f(x) = a(x-h)^2 + k$, Scheitel $S(h,k)$ |
| **Lösungsformel** | $x_{1,2} = \frac{-b \pm \sqrt{b^2-4ac}}{2a}$, $D = b^2-4ac$  |
| **Diskriminante** | $D>0$: 2 NS, $D=0$: 1 NS, $D<0$: 0 NS |
