Funktionen sind eines der zentralen Konzepte in der Mathematik. Sie beschreiben, wie eine Größe (die abhängige Variable) von einer anderen Größe (der unabhängigen Variable) abhängt. In diesem Abschnitt werden wir die Grundlagen des Funktionsbegriffs behandeln, einschließlich der Definition von Funktionen, ihrer Eigenschaften und verschiedenen Darstellungsformen.

### 1. Was ist eine Funktion?

Eine **Funktion** $f$ ist eine Zuordnung, die jedem Element $x$ aus einer Menge $D$ **genau ein** Element $y$ aus einer Menge $W$ zuordnet.

$$f: D \to W, \quad x \mapsto f(x) = y$$

| Begriff | Bedeutung |
|---------|----------|
| $x$ | Argument / unabhängige Variable |
| $f(x) = y$ | Funktionswert / abhängige Variable |
| $D = D_f$ | Definitionsbereich - alle zulässigen $x$-Werte |
| $W = W_f$ | Wertebereich - alle tatsächlich angenommenen $y$-Werte |

**Entscheidend:** Jedem $x$ wird *genau ein* $y$ zugeordnet. Umgekehrt darf aber ein $y$-Wert von mehreren $x$-Werten erreicht werden.

**Gegenbeispiel:** Die Gleichung $x^2 + y^2 = 1$ (Kreis) ist *keine* Funktion, da zu manchen $x$-Werten zwei $y$-Werte gehören.

---

### Darstellungsformen

Funktionen lassen sich auf verschiedene Arten beschreiben:

| Form | Beispiel |
|------|----------|
| **Funktionsgleichung** | $f(x) = 2x + 1$ |
| **Wertetabelle** | $x = 0 \to 1,\; x = 1 \to 3,\; x = 2 \to 5$ |
| **Graph** | Punkte $(x, f(x))$ im Koordinatensystem |
| **Verbal** | „Die Temperatur steigt pro Stunde um 2°C" |
| **Mengenschreibweise** | $f = \{(0, 1),\; (1, 3),\; (2, 5)\}$ |

### 2. Definitionsbereich und Wertebereich

Der **Definitionsbereich** $D_f$ enthält alle $x$-Werte, für die $f(x)$ definiert ist.  
Der **Wertebereich** $W_f$ enthält alle $y$-Werte, die $f$ tatsächlich annimmt.

**Typische Einschränkungen des Definitionsbereichs:**

| Situation | Bedingung | Beispiel |
|-----------|-----------|----------|
| Division durch 0 | Nenner $\ne 0$ | $f(x) = \frac{1}{x}$: $D = \mathbb{R} \setminus \{0\}$ |
| Wurzel aus negativer Zahl | Radikand $\ge 0$ | $f(x) = \sqrt{x}$: $D = [0, \infty)$ |
| Logarithmus | Argument $> 0$ | $f(x) = \ln(x)$: $D = (0, \infty)$ |
| Kombination | Beide Bedingungen | $f(x) = \frac{1}{\sqrt{x-2}}$: $D = (2, \infty)$ |

**Beispiel:** $f(x) = \sqrt{4 - x^2}$
- Bedingung: $4 - x^2 \ge 0 \Rightarrow x^2 \le 4 \Rightarrow -2 \le x \le 2$
- $D_f = [-2, 2]$, $W_f = [0, 2]$

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Definitions- und Wertebereich visualisieren ---

def visualisiere_def_wertebereich():
    """Zeigt Definitionsbereich (x-Achse) und Wertebereich (y-Achse) grafisch."""
    fig, axes = plt.subplots(1, 3, figsize=(16, 5))

    # --- f(x) = √(4-x²) ---
    ax = axes[0]
    x = np.linspace(-2, 2, 300)
    y = np.sqrt(4 - x**2)
    ax.plot(x, y, 'b-', linewidth=2.5)
    ax.axhspan(0, 2, alpha=0.1, color='green', label=r"$W_f = [0, 2]$")
    ax.axvspan(-2, 2, alpha=0.1, color='blue', label=r"$D_f = [-2, 2]$")
    ax.set_title(r"$f(x) = \sqrt{4 - x^2}$", fontsize=13)
    ax.legend(fontsize=10)

    # --- f(x) = 1/x ---
    ax = axes[1]
    x_l = np.linspace(-4, -0.1, 200)
    x_r = np.linspace(0.1, 4, 200)
    ax.plot(x_l, 1/x_l, 'b-', linewidth=2.5)
    ax.plot(x_r, 1/x_r, 'b-', linewidth=2.5)
    ax.axvline(0, color='red', linestyle=':', linewidth=1.5, label=r"$x=0$ nicht definiert")
    ax.set_title(r"$f(x) = \frac{1}{x}$, $D = \mathbb{R} \setminus \{0\}$", fontsize=13)
    ax.set_ylim(-5, 5)
    ax.legend(fontsize=10)

    # --- f(x) = ln(x) ---
    ax = axes[2]
    x = np.linspace(0.01, 6, 300)
    ax.plot(x, np.log(x), 'b-', linewidth=2.5)
    ax.axvspan(0, 0, alpha=0.3, color='red')
    ax.axvline(0, color='red', linestyle=':', linewidth=1.5, label=r"$x \leq 0$ nicht definiert")
    ax.set_title(r"$f(x) = \ln(x)$, $D = (0, \infty)$", fontsize=13)
    ax.legend(fontsize=10)

    for ax in axes:
        ax.grid(True, alpha=0.3)
        ax.axhline(0, color='black', linewidth=0.5)
        ax.axvline(0, color='black', linewidth=0.5)
        ax.set_xlabel("x")
        ax.set_ylabel("f(x)")

    plt.tight_layout()
    plt.show()

visualisiere_def_wertebereich()
```

### 3. Eigenschaften von Funktionen

| Eigenschaft | Definition | Beispiel |
|-------------|------------|----------|
| **Nullstellen** | $f(x) = 0$ | $f(x) = x^2 - 4$: Nullstellen bei $x = \pm 2$ |
| **Monoton steigend** | $x_1 < x_2 \Rightarrow f(x_1) \le f(x_2)$ | $f(x) = x^3$ auf $\mathbb{R}$ |
| **Monoton fallend** | $x_1 < x_2 \Rightarrow f(x_1) \ge f(x_2)$ | $f(x) = -x$ auf $\mathbb{R}$ |
| **Beschränkt** | $f(x)$ nimmt nicht beliebig große Werte an | $f(x) = \sin(x)$: $W = [-1, 1]$ |
| **Periodisch** | $f(x + p) = f(x)$ für alle $x$ | $\sin(x)$ mit Periode $p = 2\pi$ |

---

### Symmetrie

| Typ | Bedingung | Grafisch | Beispiel |
|-----|-----------|----------|----------|
| **Gerade** (achsensymmetrisch) | $f(-x) = f(x)$ | Spiegelung an y-Achse | $x^2, \cos(x)$ |
| **Ungerade** (punktsymmetrisch) | $f(-x) = -f(x)$ | Drehung um Ursprung (180°) | $x^3, \sin(x)$ |
| **Weder noch** | Keine der beiden | Kein Symmetriemuster | $e^x, x^2 + x$ |

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Symmetrie und Monotonie ---

def visualisiere_eigenschaften():
    """Zeigt gerade/ungerade Symmetrie und Monotonie."""
    fig, axes = plt.subplots(1, 3, figsize=(16, 5))
    x = np.linspace(-3, 3, 300)

    # --- Gerade Funktion ---
    ax = axes[0]
    ax.plot(x, x**2, 'b-', linewidth=2.5, label=r"$f(x) = x^2$")
    ax.plot(-x, x**2, 'r--', linewidth=1.5, alpha=0.5, label=r"$f(-x) = x^2$ ✓")
    ax.axvline(0, color='green', linewidth=1.5, linestyle=':', label="Symmetrieachse")
    ax.set_title("Gerade Funktion\n" + r"$f(-x) = f(x)$", fontsize=13)
    ax.legend(fontsize=10)

    # --- Ungerade Funktion ---
    ax = axes[1]
    ax.plot(x, x**3, 'b-', linewidth=2.5, label=r"$f(x) = x^3$")
    # Punktspiegelung illustrieren
    ax.scatter([1, -1], [1, -1], color='red', s=100, zorder=5)
    ax.annotate("", xy=(-1, -1), xytext=(1, 1),
                arrowprops=dict(arrowstyle='<->', color='red', lw=1.5))
    ax.scatter(0, 0, color='green', s=120, zorder=5, marker='o')
    ax.set_title("Ungerade Funktion\n" + r"$f(-x) = -f(x)$", fontsize=13)
    ax.legend(fontsize=10)

    # --- Monotonie ---
    ax = axes[2]
    x2 = np.linspace(0, 4, 300)
    ax.plot(x2, np.sqrt(x2), 'b-', linewidth=2.5, label=r"$\sqrt{x}$ - monoton steigend")
    ax.plot(x2, 2 - 0.5*x2, 'r-', linewidth=2.5, label=r"$2 - 0{,}5x$ - monoton fallend")
    ax.set_title("Monotonie", fontsize=13)
    ax.legend(fontsize=10)

    for ax in axes:
        ax.grid(True, alpha=0.3)
        ax.axhline(0, color='black', linewidth=0.5)
        ax.axvline(0, color='black', linewidth=0.5)
        ax.set_xlabel("x")
        ax.set_ylabel("f(x)")

    plt.tight_layout()
    plt.show()

visualisiere_eigenschaften()
```

### 4. Umkehrfunktion

Die **Umkehrfunktion** $f^{-1}$ kehrt die Zuordnung um: Statt $x \mapsto y$ wird $y \mapsto x$.

$$f^{-1}(f(x)) = x \quad \text{und} \quad f(f^{-1}(y)) = y$$

**Voraussetzung:** $f$ muss **injektiv** (eineindeutig) sein - jeder $y$-Wert wird höchstens einmal angenommen. Das ist genau dann der Fall, wenn $f$ **streng monoton** ist.

**Bestimmung der Umkehrfunktion:**
1. Schreibe $y = f(x)$
2. Löse nach $x$ auf
3. Vertausche $x$ und $y$

**Beispiel:** $f(x) = 2x + 3$
1. $y = 2x + 3$
2. $x = \frac{y - 3}{2}$
3. $f^{-1}(x) = \frac{x - 3}{2}$

**Grafisch:** Der Graph von $f^{-1}$ entsteht durch Spiegelung des Graphen von $f$ an der Geraden $y = x$.

| Funktion | Umkehrfunktion |
|----------|----------------|
| $f(x) = x^2$ (für $x \ge 0$) | $f^{-1}(x) = \sqrt{x}$ |
| $f(x) = e^x$ | $f^{-1}(x) = \ln(x)$ |
| $f(x) = \sin(x)$ (für $x \in [-\frac{\pi}{2}, \frac{\pi}{2}]$) | $f^{-1}(x) = \arcsin(x)$ |

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Umkehrfunktion visualisieren ---

def visualisiere_umkehrfunktion():
    """Zeigt Funktion, Umkehrfunktion und die Spiegelachse y=x."""
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))

    # --- f(x) = x² und f⁻¹(x) = √x ---
    ax = axes[0]
    x = np.linspace(0, 3, 300)
    ax.plot(x, x**2, 'b-', linewidth=2.5, label=r"$f(x) = x^2$")
    ax.plot(x, np.sqrt(x), 'r-', linewidth=2.5, label=r"$f^{-1}(x) = \sqrt{x}$")
    diag = np.linspace(0, 3, 100)
    ax.plot(diag, diag, 'k--', alpha=0.4, label=r"$y = x$")
    # Spiegelpunkte
    ax.scatter([2, 4], [4, 2], color='green', s=100, zorder=5)
    ax.annotate("(2, 4)", xy=(2, 4), xytext=(2.3, 4.5), fontsize=11)
    ax.annotate("(4, 2)", xy=(4, 2), xytext=(4.3, 2.5), fontsize=11)
    ax.set_title(r"$x^2$ und $\sqrt{x}$ (für $x \geq 0$)", fontsize=13)
    ax.legend(fontsize=11)
    ax.set_xlim(-0.5, 5)
    ax.set_ylim(-0.5, 5)
    ax.set_aspect('equal')

    # --- f(x) = eˣ und f⁻¹(x) = ln(x) ---
    ax = axes[1]
    x1 = np.linspace(-2, 2, 300)
    x2 = np.linspace(0.05, 7, 300)
    ax.plot(x1, np.exp(x1), 'b-', linewidth=2.5, label=r"$f(x) = e^x$")
    ax.plot(x2, np.log(x2), 'r-', linewidth=2.5, label=r"$f^{-1}(x) = \ln(x)$")
    diag = np.linspace(-2, 7, 100)
    ax.plot(diag, diag, 'k--', alpha=0.4, label=r"$y = x$")
    ax.set_title(r"$e^x$ und $\ln(x)$", fontsize=13)
    ax.legend(fontsize=11)
    ax.set_xlim(-2.5, 7)
    ax.set_ylim(-2.5, 7)
    ax.set_aspect('equal')

    for ax in axes:
        ax.grid(True, alpha=0.3)
        ax.axhline(0, color='black', linewidth=0.5)
        ax.axvline(0, color='black', linewidth=0.5)
        ax.set_xlabel("x")
        ax.set_ylabel("y")

    plt.tight_layout()
    plt.show()

visualisiere_umkehrfunktion()
```

## Verständnis-Check

*Löse die folgenden Aufgaben auf Papier.*

**Aufgabe 1:** Bestimme den Definitionsbereich von $f(x) = \dfrac{\sqrt{x+3}}{x-1}$.

**Aufgabe 2:** Ist $f(x) = x^4 - 2x^2$ eine gerade, ungerade oder weder-noch-Funktion?

**Aufgabe 3:** Bestimme die Umkehrfunktion von $f(x) = 3x - 7$.

**Aufgabe 4:** Eine Funktion $f$ hat den Definitionsbereich $D = [-2, 5]$ und den Wertebereich $W = [0, 9]$. Welchen Definitions- und Wertebereich hat $f^{-1}$?

### Lösungen

<details>
<summary>Lösung Aufgabe 1</summary>

Zwei Bedingungen:
1. Wurzel: $x + 3 \ge 0 \Rightarrow x \ge -3$
2. Nenner: $x \ne 1$

$$D_f = [-3, \infty) \setminus \{1\} = [-3, 1) \cup (1, \infty)$$
</details>

<details>
<summary>Lösung Aufgabe 2</summary>

Prüfe $f(-x)$:
$$f(-x) = (-x)^4 - 2(-x)^2 = x^4 - 2x^2 = f(x)$$

$f(-x) = f(x)$ → **gerade Funktion** (achsensymmetrisch zur y-Achse).
</details>

<details>
<summary>Lösung Aufgabe 3</summary>

1. $y = 3x - 7$
2. $x = \frac{y + 7}{3}$
3. $f^{-1}(x) = \frac{x + 7}{3}$

Probe: $f(f^{-1}(x)) = 3 \cdot \frac{x+7}{3} - 7 = x$ ✓
</details>

<details>
<summary>Lösung Aufgabe 4</summary>

Bei der Umkehrfunktion werden Definitions- und Wertebereich vertauscht:
$$D_{f^{-1}} = W_f = [0, 9] \qquad W_{f^{-1}} = D_f = [-2, 5]$$
</details>

## Zusammenfassung

| Konzept | Kernaussage |
|---------|-------------|
| **Funktion** | Zuordnung $x \mapsto f(x)$, jedem $x$ genau ein $y$ |
| **Definitionsbereich** | Alle zulässigen $x$-Werte |
| **Wertebereich** | Alle tatsächlich angenommenen $y$-Werte |
| **Vertikale-Linie-Test** | Graph ist Funktion ⟺ jede Vertikale schneidet höchstens einmal |
| **Monotonie** | Steigend/fallend - Verhalten des Funktionswerts |
| **Symmetrie** | Gerade: $f(-x)=f(x)$, Ungerade: $f(-x)=-f(x)$ |
| **Umkehrfunktion** | Spiegelung an $y=x$, existiert bei strenger Monotonie |
