# Vorwort

Dieses Dokument ist eine **konsolidierte, programmatisch erzeugte Sammlung** von Jupyter-Notebooks. Es ist ein Versuch möglichst viel Mathematische Theorie in möglichst kompakter Form darzustellen. 
Dazu kommen Visualisierungen, Beispiele und gelegentlich Aufgaben mit Lösungen. Obwohl alle Kapitel "Lernziele" beschreiben, ist das Ziel Klarheit und Nachschlagbarkeit, möglicherweise nicht der erste Zugang (Gerne zuerst tiefer in die Themen gehen, dann hier nachschlagen).

## Über die Formatierung dieses Dokuments

Dieses Dokument entstand durch Export der einzelnen `.ipynb`-Dateien und anschließende Konsolidierung. Das führt zu Eigenheiten:

- **Seitenumbrüche** sind teilweise unglücklich platziert — etwa mitten in Tabellen oder direkt nach einer Überschrift. Das ist kein Fehler in deinem PDF-Reader, sondern eine Grenze meiner Zeit die Konvertierung zu verbessern.
- **Abbildungen und Plots** werden in einer abgeleiteten Größe eingefügt; an manchen Stellen können sie quetschen oder überquellen (vorallem Mehr-Diagramm-Bilder haben derzeit Skalierungsprobleme; wer möchte, kann sie im [Export\Markdown](https://github.com/Officialstjp/python-math/tree/exports/export/short/Foundations/markdown)-Pfad im GitHub-Repository finden).
- **Formatierungen** wie farbige Boxen oder spezielle Layouts funktionieren nicht in allen PDF-Readern gleich.

Ein manuelles Dokument (z.B. in Word oder LaTeX) hätte glattere Übergänge, hier steht mir schnelle Generierbarkeit jedoch über perfekter Typografie.

**Die Originalnotebooks selbst sind fehlerfrei** und lassen sich einzeln problemlos lesen. Alle lassen sich ebenso im [Repository](https://github.com/Officialstjp/python-math) finden und dort auch einzeln in allen Formaten einsehen. 

Für die Ausführung der Notebooks und deren Python-Code und Export / Konsolidierung der Notebooks ist eine lokale Python-Umgebung erforderlich.

Checke auch gerne die [Machine-Learning Einführung](https://github.com/Officialstjp/python-math/blob/exports/export/short/ml/pdf/MachineLearning_Math.pdf) aus, die auf diesen Stoff aufbaut. Die selben Prinizipin sind treffen dort auch zu.

## Struktur und Umfang

Das Material gliedert sich in **neun Hauptkapitel** (von Zahlenmengen bis Statistik) sowie Glossare. Jedes Notebook ist in sich geschlossen, verweist aber auf Voraussetzungen und verwandte Themen in anderen Kapiteln.

Alle leiten in den ersten Zeilen kürz über den **Umfang des Kapitels** und **die Lernziele** (/ Kompetenzen) ein. 

Zwischen textuellen Beschreibungen sind immer wieder **Visualisierungen** und **Verifizierungen** der Themen per Python (Ausgaben) eingebettet. (Für Source-Code, siehe [Original-Notebooks](https://github.com/Officialstjp/python-math/tree/master/Foundations) )

Alle Kapitel enden mit Kompetenz-Checks (Aufgaben mit Lösungen), Zusammenfassungen und Deutsch/Englisch-Begriffs-Glossaren. (In Kurzversionen sind die **Code-Outputs, Glossare** und **Kompetenz-Checks** **ausgelassen**, um die Seitenzahl zu reduzieren.)

Die ersten Kapitel verwenden generell Tabellen für Zusammenfassungen, ab Statistik dann mehr Fließtext. Das ist teilweise eine Frage der Themen, teilweise aber auch mal wieder, meiner Zeit.
