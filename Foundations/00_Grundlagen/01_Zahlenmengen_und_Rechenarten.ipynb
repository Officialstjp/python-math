## 01 - Zahlenmengen und Grundrechenarten

Die hierarchischen Zahlenmengen $\mathbb{N} \subset \mathbb{Z} \subset \mathbb{Q} \subset \mathbb{R}$ ($\subset$ = "ist Teilmenge von", kommt gleich) und die Grundrechenarten bilden die Grundlage für alle weiteren mathematischen Konzepte.

## 1. Die Zahlenmengen

Die Mathematik organisiert Zahlen in hierarchischen Mengen, die aufeinander aufbauen:

### $\mathbb{N}$ - Natürliche Zahlen
Die "Zählzahlen" beginnen bei 1 und gehen unendlich weiter:
$$\mathbb{N} = \{1, 2, 3, 4, \ldots\}$$

Je nach Konvention gehört die $0$ dazu ($\mathbb{N}_0$) oder nicht. Wir verwenden $\mathbb{N}_0 = \{0, 1, 2, 3, 4, \ldots\}$.

**Eigenschaften:** 
- Addition und Multiplikation bleiben immer in $\mathbb{N}$. 
- Subtraktion und Division hingegen nicht - $3 - 5 = -2$ liegt außerhalb von $\mathbb{N}$, und $1 \div 3$ ist kein ganzzahliges Ergebnis. 

### $\mathbb{Z}$ - Ganze Zahlen
Natürliche Zahlen erweitern wir um negative Zahlen und die Null:
$$\mathbb{Z} = \{\ldots, -3, -2, -1, 0, 1, 2, 3, \ldots\}$$

**Eigenschaften:**
- Subtraktion immer möglich: $3 - 5 = -2 \in \mathbb{Z}$.
- Abgeschlossen unter Addition, Subtraktion und Multiplikation. **Nicht** abgeschlossen unter Division.



### $\mathbb{Q}$ - Rationale Zahlen
Alle Zahlen, die sich als **Bruch** zweier ganzer Zahlen darstellen lassen:
$$\mathbb{Q} = \left\{ \frac{p}{q} \mid p \in \mathbb{Z}, q \in \mathbb{Z}, q \neq 0 \right\}$$

**Beispiele:** 
- $\frac{1}{2}, \frac{-3}{4}, \frac{7}{1} = 7$ 
- $\quad 0.333\ldots = \frac{1}{3}$

**Eigenschaften:**
- Division wird vollständig möglich (bis auf Division durch Null).
- Dezimaldarstellung ist entweder endlich oder periodisch (wiederkehrendes Muster).

Aber: $\sqrt{2} \notin \mathbb{Q}$ - es gibt keine ganzen Zahlen $p, q$ mit $\frac{p}{q} = \sqrt{2}$.

### $\mathbb{R}$ - Reelle Zahlen
Alle Punkte auf dem Zahlenstrahl, inklusive irrationaler Zahlen:
$$\mathbb{R} = \mathbb{Q} \cup \{\text{irrationale Zahlen}\}$$

**Irrationale Zahlen:** $\sqrt{2}, \pi, e, \sqrt{3}, \ldots$ (unendlich, nicht periodisch)

### Die Hierarchie der Zahlenmengen
$$\mathbb{N} \subset \mathbb{Z} \subset \mathbb{Q} \subset \mathbb{R}$$

*Jede natürliche Zahl ist auch eine ganze Zahl, jede ganze Zahl ist eine rationale Zahl und jede rationale Zahl ist eine reelle Zahl.*

```python
# === KONFIGURATION: Visualisierung der Zahlenmengen ===
Zahlenmengen_Config = {
    "natuerlich": [1, 2, 3, 5, 8, 13, 21],
    "ganz": [-5, -3, -1, 0, 1, 3, 5],
    "rational": [-2.5, -1.0, 0.0, 0.5, 1.3333, 2.75],
    "reell": [1.4142, 3.1415, 2.7182, 0.5772], # Wurzel(2), Pi, e, Euler-Mascheroni-Konstante (Fun Fact: zurzeit nicht bekannt, ob diese Zahl rational oder irrational ist)

    # Darstellung
    "figsize": (6, 4),
    "zahlenstrahl_bereich": (-5, 5),
    "farben": {
        "N": "#2ebc71",
        "Z": "#3488db",
        "Q": "#9b59b7",
        "R": "#e74c3f"
    },
    "marker_size": 100
}
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

def plot_zahlenmengen_venn(cfg):
    """Visualisiert die Hierarchie der Zahlenmengen als verschaltelte Mengen."""
    fig, ax = plt.subplots(figsize=cfg["figsize"])
    ax.set_xlim(-3.5, 5)
    ax.set_ylim(-2.5, 2.5)
    ax.set_aspect('equal')
    ax.axis('off')
    ax.set_title("Die Hierarchie der Zahlenmengen", fontsize=16, fontweight='bold')

    # Verschachtelte Ellipsen (von außen nach innen)
    ellipsen = [
        (0, 0, 2.8, 2.2, cfg["farben"]["R"], r"$\mathbb{R}$ (Reelle Zahlen)", 0.2),
        (0, 0, 2.2, 1.7, cfg["farben"]["Q"], r"$\mathbb{Q}$ (Rationale)", 0.3),
        (0, 0, 1.5, 1.2, cfg["farben"]["Z"], r"$\mathbb{Z}$ (Ganze)", 0.4),
        (0, 0, 0.8, 0.6, cfg["farben"]["N"], r"$\mathbb{N}$ (Natürliche)", 0.5),
    ]

    for x, y, w, h, color, label, alpha in ellipsen:
        ellipse = patches.Ellipse((x, y), w*2, h*2, facecolor=color,
                                   edgecolor='black', alpha=alpha, lw=2)
        ax.add_patch(ellipse)

    # Beschriftungen
    ax.text(0, 0, r"$\mathbb{N}$", ha='center', va='center', fontsize=18, fontweight='bold')
    ax.text(0, 0.9, r"$\mathbb{Z}$", ha='center', va='center', fontsize=16)
    ax.text(0, 1.45, r"$\mathbb{Q}$", ha='center', va='center', fontsize=16)
    ax.text(0, 2.0, r"$\mathbb{R}$", ha='center', va='center', fontsize=16)

    # Beispiele rechts
    beispiele = [
        (3.5, 1.8, r"$\pi, \sqrt{2}, e, \gamma$", cfg["farben"]["R"]),
        (3.5, 1.2, r"$\frac{1}{2}, 0.\dot{3}$", cfg["farben"]["Q"]),
        (3.5, 0.6, r"$-3, -1, 0$", cfg["farben"]["Z"]),
        (3.5, 0, r"$1, 2, 3, ...$", cfg["farben"]["N"]),
    ]

    for x, y, text, color in beispiele:
        ax.text(x, y, text, fontsize=12, color='black',
                bbox=dict(boxstyle='round', facecolor=color, alpha=0.3))

    return fig

fig = plot_zahlenmengen_venn(Zahlenmengen_Config)
plt.tight_layout()
plt.show()
```

```python
def plot_zahlenstrahl(cfg):
    """Visualisiert verschieden Zahlentypen auf dem Zahlenstrahl."""
    fig, ax = plt.subplots(figsize=(8, 4))

    x_min, x_max = cfg["zahlenstrahl_bereich"]
    ax.set_xlim(x_min - 0.5, x_max + 0.5)
    ax.set_ylim(-1, 2)
    ax.axis('off')
    ax.set_title("Zahlen auf dem Zahlenstrahl", fontsize=14, fontweight='bold')

    # Zahlenstrahl zeichnen
    ax.annotate('', xy=(x_max + 0.3, 0), xytext=(x_min - 0.3, 0),
                arrowprops=dict(arrowstyle='->', lw=2, color='black'))

    # Markierungen
    for i in range(x_min, x_max + 1):
        ax.plot([i, i], [-0.1, 0.1], 'k-', lw=1)
        ax.text(i, -0.3, str(i), ha='center', va='top', fontsize=10)

    # Natürliche Zahlen
    for n in cfg["natuerlich"]:
        if x_min <= n <= x_max:
            ax.scatter(n, 0.25, c=cfg["farben"]["N"], s=cfg["marker_size"],
                       zorder=5, edgecolor='black', label='ℕ' if n == cfg["natuerlich"][0] else '')

    # Ganze Zahlen
    for z in cfg["ganz"]:
        if x_min <= z <= x_max:
            ax.scatter(z, 0.5, c=cfg["farben"]["Z"], s=cfg["marker_size"],
                      zorder=5, edgecolor='black', marker='s')

    # Rationale Zahlen
    for q in cfg["rational"]:
        if x_min <= q <= x_max:
            ax.scatter(q, 0.75, c=cfg["farben"]["Q"], s=cfg["marker_size"]*0.7,
                      zorder=5, edgecolor='black', marker='^')

    # Reelle Zahlen
    for p in cfg["reell"]:
        if x_min <= p <= x_max:
            ax.scatter(p, 1.0, c=cfg["farben"]["R"], s=cfg["marker_size"],
                       zorder=5, edgecolor='black', label='o')

    # Legende
    from matplotlib.lines import Line2D
    legend_elements = [
        Line2D([0], [0], marker='o', color='w', markerfacecolor=cfg["farben"]["N"],
               markersize=10, label='Natürlich (ℕ)'),
        Line2D([0], [0], marker='s', color='w', markerfacecolor=cfg["farben"]["Z"],
               markersize=10, label='Ganz (ℤ)'),
        Line2D([0], [0], marker='^', color='w', markerfacecolor=cfg["farben"]["Q"],
               markersize=10, label='Rational (ℚ)'),
        Line2D([0], [0], marker='o', color='w', markerfacecolor=cfg["farben"]["R"],
               markersize=10, label='Reell (irrational) (ℝ)'),
    ]
    ax.legend(handles=legend_elements, loc='upper right')

    return fig

fig = plot_zahlenstrahl(Zahlenmengen_Config)
plt.tight_layout()
plt.show()
```

### Exakte vs. numerische Darstellung

| Art | Bedeutung | Beispiel |
|-----|-----------|----------|
| **Exakt** | Wert als Symbol oder Bruch - kein Runden | $\frac{1}{3}, \quad \sqrt{2}, \quad \pi$ |
| **Numerisch** | Dezimalzahl mit endlich vielen Stellen - immer eine Näherung | $0{,}333, \quad 1{,}4142, \quad 3{,}14159$ |

In der Mathematik ist $\sqrt{2}$ exakt - $1{,}4142$ ist bereits eine Näherung. Taschenrechner und Computer arbeiten fast immer numerisch, was zu Rundungsfehlern führen kann. Faustregel: Solange du rechnen kannst, bleib exakt. Erst am Ende runden.

## 2. Die Grundrechenarten und ihre Gesetze

Die vier Grundrechenarten sind: 
- Addition (+), 
- Subtraktion (-), 
- Multiplikation (·), 
- Division (÷). 

Drei strukturelle Gesetze beschreiben, wie Terme umgeformt werden können und zu wissen, für welche Operationen sie *nicht* gelten, schützt vor den häufigsten Fehlern.

#### A. Kommutativgesetz (Vertauschungsgesetz)
Die **Reihenfolge der Operanden** ändert das Ergebnis bei Addition und Multiplikation **nicht**:
$$a + b = b + a \quad \text{und} \quad a \cdot b = b \cdot a$$

**! Gilt NICHT für Subtraktion und Division:**
$$5 - 3 \neq 3 - 5 \quad \text{und} \quad 6 \div 2 \neq 2 \div 6$$

#### B. Assoziativgesetz (Verbindungsgesetz)
Klammern dürfen **verschoben** werden:
$$(a + b) + c = a + (b + c) \quad \text{und} \quad (a \cdot b) \cdot c = a \cdot (b \cdot c)$$

**! Gilt NICHT für Subtraktion und Division:**
$$(8 - 4) - 2 = 2 \quad \text{aber} \quad 8 - (4 - 2) = 6$$

#### C. Distributivgesetz (Verteilungsgesetz)
Multiplikation wird über Addition/Subtraktion **"verteilt"**:
$$a \cdot (b + c) = a \cdot b + a \cdot c$$

**Beispiel:** $3 \cdot (4 + 5) = 3 \cdot 4 + 3 \cdot 5 = 12 + 15 = 27$

Das Distributivgesetz ist die Grundlage für das Ausmultiplizieren von Klammern.

#### D. Rechenreihenfolge (PEMDAS / Punkt vor Strich)
1. **P**arentheses (Klammern)
2. **E**xponents (Potenzen)
3. **M**ultiplication & **D**ivision (von links nach rechts)
4. **A**ddition & **S**ubtraction (von links nach rechts)

**Beispiel:** $2 + 3 \cdot 4^2 = 2 + 3 \cdot 16 = 2 + 48 = 50$

```python
# === KONFIGURATION: Rechengesetze Beispiele ===
Rechengesetze_Config = {
    # Beispielwerte für Demonstrationen
    "a": 5,
    "b": 3,
    "c": 2,

    # Komplexere Ausdrücke für Reihenfolge
    "ausdruecke": [
        ("2 + 3 * 4", 14),          # Punkt vor Strich
        ("(2 + 3) * 4", 20),        # Klammern zuerst
        ("2 + 3 * 4**2", 50),       # Potenz vor Multiplikation
        ("24 / 4 / 2", 3),          # Linksassoziativität
        ("24 / (4 / 2)", 12),       # Klammern ändern Reihenfolge
    ],

    # Typische Fehler
    "fehler_beispiele": [
        ("(-3)**2 vs -3**2", "9 vs -9", "Klammern bei negativen Basen!"),
        ("8 - 4 - 2", "2 (nicht 6)", "Links nach rechts auswerten"),
    ]
}
```

```python
def demonstriere_rechengesetze(cfg):
    """Verifiziert die Rechengesetze mit konkreten Zahlen."""
    a, b, c = cfg["a"], cfg["b"], cfg["c"]

    print("=" * 50)
    print(f"Demonstration mit a={a}, b={b}, c={c}")
    print("=" * 50)

    # Kommutativgesetz
    print("\n--- KOMMUTATIVGESETZ ---")
    print(f"  Addition:       {a} + {b} = {a+b}  |  {b} + {a} = {b+a}  ✓" if a+b == b+a else "")
    print(f"  Multiplikation: {a} · {b} = {a*b}  |  {b} · {a} = {b*a}  ✓" if a*b == b*a else "")
    print(f"  Subtraktion:    {a} - {b} = {a-b}  |  {b} - {a} = {b-a}  ✗ (nicht kommutativ!)")

    # Assoziativgesetz
    print("\n--- ASSOZIATIVGESETZ ---")
    print(f"  ({a}+{b})+{c} = {(a+b)+c}  |  {a}+({b}+{c}) = {a+(b+c)}  ✓")
    print(f"  ({a}·{b})·{c} = {(a*b)*c}  |  {a}·({b}·{c}) = {a*(b*c)}  ✓")
    print(f"  ({a}-{b})-{c} = {(a-b)-c}  |  {a}-({b}-{c}) = {a-(b-c)}  ✗")

    # Distributivgesetz
    print("\n--- DISTRIBUTIVGESETZ ---")
    links = a * (b + c)
    rechts = a * b + a * c
    print(f"  {a} · ({b} + {c}) = {links}")
    print(f"  {a}·{b} + {a}·{c} = {rechts}")
    print(f"  Beide gleich? {'✓' if links == rechts else '✗'}")

    # Rechenreihenfolge
    print("\n--- RECHENREIHENFOLGE ---")
    for ausdruck, erwartet in cfg["ausdruecke"]:
        ergebnis = eval(ausdruck)
        status = "✓" if ergebnis == erwartet else "✗"
        print(f"  {ausdruck:20} = {ergebnis:6} {status}")

demonstriere_rechengesetze(Rechengesetze_Config)
```

## Zusammenfassung

| Konzept | Kernaussage |
|---------|-------------|
| Zahlenmengen | $\mathbb{N} \subset \mathbb{Z} \subset \mathbb{Q} \subset \mathbb{R}$ - jede Menge erweitert die vorherige |
| Natürliche Zahlen $\mathbb{N}$ | Zählzahlen ab 1 (bzw. 0 bei $\mathbb{N}_0$) |
| Ganze Zahlen $\mathbb{Z}$ | $\mathbb{N}$ erweitert um negative Zahlen und Null |
| Rationale Zahlen $\mathbb{Q}$ | Alle Brüche $\frac{p}{q}$, endliche oder periodische Dezimalzahlen |
| Reelle Zahlen $\mathbb{R}$ | $\mathbb{Q}$ plus irrationale Zahlen ($\sqrt{2}, \pi, e$) |
| Rechengesetze | Kommutativ, Assoziativ, Distributiv - gelten nur für bestimmte Operationen |
| Rechenreihenfolge | Klammern → Potenzen → Punkt → Strich (PEMDAS) |
