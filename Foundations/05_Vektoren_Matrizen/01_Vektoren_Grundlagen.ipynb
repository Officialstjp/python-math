# Vektoren

Viele physikalische Größen lassen sich nicht allein durch ihren Betrag vollständig charakterisieren. Zusätzlich ist noch die Angabe einer Richtung notwendig.
Solche gerichtete Größen bezeichnet man als Vektoren.








## Definition

Ein **Vektor** beschreibt eine gerichtete Größe: **Länge** und **Richtung**.

Gemometrisch kann man ihn als **Mebe aller gleich langen, gleich gerichtiten und gleich orientierten Pfeile** auffassen, die im Raum existieren können.

Ein einzelner Pfeil dieser Menge ist ein Repräsentant (Stellvertreter) des entsprechenden Vektors.


### Notation
- Vektoren: $\vec a,\ \vec b$ (oder fett: $\mathbf a,\ \mathbf b$)
- Betrag / Norm: $|\vec a|$ oder $\|\vec a\|$

### Nullvektor
Der **Nullvektor** ist der Vektor mit Betrag $0$:
$$
\vec 0 = \begin{pmatrix}0\\0\end{pmatrix}\quad (\text{bzw. } \begin{pmatrix}0\\0\\0\end{pmatrix})
$$
Er hat **keine Richtung** (bzw. Richtung ist nicht sinnvoll definiert).

### Einheitsvektor
Ein **Einheitsvektor** hat Betrag $1$ und zeigt in eine bestimmte Richtung.

Der Einheitsvektor in Richtung eines Vektors $\vec a \neq \vec 0$ ist:
$$
\hat a = \frac{\vec a}{\|\vec a\|}.
$$

```python
# Simple Visualisierung eines Vektors in einem Koordinatensystem

import matplotlib.pyplot as plt
import numpy as np

# Grid definieren
fig, ax = plt.subplots()
ax.set_xlim(-5, 5)
ax.set_ylim(-5, 5)

ax.grid(True)

# Vektor definieren
origin = np.array([0, 0])  # Ursprung
vector = np.array([3, 4])  # Vektor-Komponenten

# Vektor zeichnen
ax.quiver(*origin, *vector, angles='xy', scale_units='xy', scale=1, color='r')
ax.set_aspect('equal')
ax.set_title('Vektor in einem Koordinatensystem')

# Gegenvektor zeichnen
ax.quiver(*origin, *-vector, angles='xy', scale_units='xy', scale=1, color='b', alpha=0.5)
ax.legend(['Vektor', 'Gegenvektor'], loc='upper left')

# Beschriftungen hinzufügen
ax.text(vector[0]/2 - 1, vector[1]/2, 'Vektor', color='r', fontsize=12, ha='center')
ax.text(-vector[0]/2 + 1.7, -vector[1]/2, 'Gegenvektor', color='b', fontsize=12, ha='center')


plt.show()
```

# Regeln für Vektoren

- Der Vektor $\vec{BA} = -\vec{AB}$ heißt **Gegenvektor** des Vektors $\vec{AB}$.
- Der Gegenvektor $-\vec a$ hat **gleichen Betrag**, **gleiche Richtung**, aber **entgegengesetzte Orientierung**.

## Addition und Subtraktion 

- Vektoren werden komponentenweise **addiert** bzw. **subtrahiert**.
$$
\vec{a} \pm \vec{b} 
= 
\begin{pmatrix} a_x \\ a_y \end{pmatrix} 
\pm 
\begin{pmatrix} b_x \\ b_y \end{pmatrix}
= 
\begin{pmatrix} a_x \pm b_x \\ a_y \pm b_y \end{pmatrix}
$$

**Grafisch:** Bei $\vec a + \vec b$ wird der **Schaft** von $\vec b$ an die **Spitze** von $\vec a$ gesetzt.  
Die Summe ist der Pfeil **vom Schaft von $\vec a$ zur Spitze von $\vec b$** und wird als Vektorsumme bezeichnet.


## Skalarmultiplikation (komponentenweise)
Für einen Skalar $c \in \mathbb{R}$ gilt:
$$
c \cdot \vec a
=
c \cdot \begin{pmatrix} a_x \\ a_y \end{pmatrix}
=
\begin{pmatrix} c\,a_x \\ c\,a_y \end{pmatrix}.
$$

- $c>0$: Richtung bleibt gleich, Länge skaliert.
- $c<0$: Orientierung kehrt sich um, Länge skaliert mit $|c|$.

## Audrückung in Kräften

Vektoren werden häufig verwendet um Physikalische Kräfte zu beschreiben.

Zwei Vektoren in einem Koordinatensystem können als "Vektor-Parallelogramm" dargestellt werden. Die Mittellinie (jene, die vom Anfangspunkt zum gegenüberliegenden Ende des Parallelogramms existiert), stellt den resultierenden Vektor dar.
