### 1. Winkel an Geradenkreuzungen

Wenn sich zwei Geraden in einer Ebene schneiden, entstehen um den Schnittpunkt vier Winkel. Diese stehen in einer festen Beziehung zueinander, welche wir verwenden können um fehlende Winkelgrößen ohne messen zu müssen.

#### A. Nebenwinkel (Supplementäre Winkel)

Nehmen wir eine gerade Linie, die von einer anderen Linie geschnitten wird. 
Zwei Winkel, die **nebeneinander** liegen, einen gemeinsamen Schenkel und einen gemeinsamen Scheitelpunkt haben, nennt man **Nebenwinkel**

Die beiden äußeren Schenkel bilden zusammen eine gerade (einen gestreckten Winkel von $180°$), daher gilt für Nebenwinkel immer:

$$ \alpha + \beta = 180° $$

Man sagt auch: "Die Winkel ergänzen sich zu $180°$ (sie sind supplementär)"
Wenn man einen Winkel kennt, kann man sofort den anderen berechnen: $\beta = 180° - \alpha$.

#### B. Scheitelwinkel

Zwei Winkel, die sich an einer Geradenkreuzung **gegenüberliegen**, nennt man **Scheitelwinkel**.
Sie teilen sich den Scheitelpunkt, haben aber keinen gemeinsamen Schenkel.

Scheitelwinkel sind **immer gleich groß**.

**Warum ist das so?**

Nehmen wir die Nebenwinkel $\alpha$ und $\beta$. Somit $\alpha + \beta = 180°$ 
Gleichzeitig sind aber auch $\beta$ und $\gamma$ Nebenwinkel. Also ist $\beta + \gamma = 180°$.
Daraus folgt logisch: $\alpha$ muss genau so groß sein wie $\gamma$, um $\beta$ zu $180°$ zu ergänzen.

Sehen wir uns das ganze visualisiert an:


```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Arc


def plot_intersection(ax):
    ax.set_xlim(-2, 2)
    ax.set_ylim(-2, 2)
    ax.set_aspect('equal')
    ax.axis('off')
    ax.set_title("Winkel an einer Geradenkreuzung", fontsize=16)

    # Zwei sich schneidende Geraden
    x = np.linspace(-1.8, 1.8, 100)
    ax.plot(x, 0.5 * x, 'k-', lw=2)

    # Gerade 2: y = -0.5x
    ax.plot(x, -0.5 * x, 'k-', lw=2)
    
    # Winkelbögen und Beschriftungen
    # Winekl Alpha (rechts)
    ax.add_patch(Arc((0,0), 1, 1, angle=0, theta1=-26.5, theta2=26.5, color='blue', lw=2))
    ax.text(0.6, 0, r'$\alpha$', fontsize=14, color='blue', va='center', ha='center')

    # Winkel Beta (oben)
    ax.add_patch(Arc((0,0), 1, 1, angle=0, theta1=26.5, theta2=153.5, color='red', lw=2))
    ax.text(0, 0.6, r'$\beta$', fontsize=14, color='red', va='center', ha='center')

    # Winkel Gamma (links) - Scheitelwinkel zu Alpha
    ax.add_patch(Arc((0,0), 1, 1, angle=0, theta1=153.5, theta2=206.5, color='blue', lw=2))
    ax.text(-0.7, 0, r'$\gamma$', fontsize=14, color='blue', va='center', ha='center')

    # Winkel Delta (unten) - Scheitelwinkel zu Beta
    ax.add_patch(Arc((0,0), 1, 1, angle=0, theta1=206.5, theta2=333.5, color='red', lw=2))
    ax.text(0, -0.7, r'$\delta$', fontsize=14, color='red', va='center', ha='center')

    # Erklärungstext
    info_text = (
        r"$\alpha = \gamma$ (Scheitelwinkel)" + "\n" +
        r"$\beta = \delta$ (Scheitelwinkel)" + "\n" +
        r"$\alpha + \beta = 180^\circ$ (Nebenwinkel)" + "\n" +
        r"$\gamma + \delta = 180^\circ$ (Nebenwinkel)"
    )
    ax.text(-1.8, -1.8, info_text, fontsize=12, bbox=dict(facecolor='white', alpha=0.8))

fig, ax = plt.subplots(figsize=(6, 6))
plot_intersection(ax)
plt.show()
```

### 2. Winkel an Parallelen

Wenn eine Gerade zwei andere Geraden schneiden, die **parallel** zueinander sind, treten zwei weitere Winkelarten auf: **F-Winkel** und **Z-Winkel** (Da die Linienführung an diese Buchstaben erinnert)
Diese Situation sieht man bspw. bei Eisenbahnschienen, Fensterrahmen oder Straßenkreuzungen.

#### A. Stufenwinkel (F-Winkel)

Schneidet man die Kreuzung der oberen Parallele aus und schiebt die schneidende Gerade (**Transversale**) hinunter, bis sie genau auf der unteren Parallele liegt, würden sich die Winkel exakt decken.
Winkel, die auf derselben Seite der Transversale und auf derselben Seite der Parallelen liegen (also: beide "oben rechts" oder beide "unten links"), nennt man **Stufenwinkel**.

**Gesetz:** An parallelen Geraden sind Stufenwinkel gleich groß.
$$ \alpha_1 = \alpha_2 $$

#### B. Wechselwinkel (Z-Winkel)
Wechselwinkel liegen auf entgegengesetzten Seiten der schneidenden Geraden. Man sie sich als Kombination der Scheitelwinkel und Stufenwinkel vorstellen.
Wenn man dem Buchstaben "Z" folgt, liegen die Winkel in den inneren Ecken des Z.

**Gesetz:** An parallelen Geraden sind Wechselwinkel gleich groß.
$$ \alpha_1 = \gamma_2 $$

Diese Gesetze gelten **nur**, wenn die geschnittenen Geraden auch wirklich parellel sind.
Umgekehrt, wenn wir gleich große Stufenwinkel finden, *müssen* die Geraden parallel sein.

```python
# Visualisierung von Stufen- und Wechselwinkeln
fig, ax = plt.subplots(figsize=(8, 6))
ax.set_xlim(-1, 6)
ax.set_ylim(-1, 4)
#ax.axis('off')
ax.set_title("Stufen- und Wechselwinkel", fontsize=16)

# Parallele Geraden
x = np.linspace(-0.5, 5.5, 100)
ax.plot(x, [1]*100, 'k_', lw=2, label='Parallel g')
ax.plot(x, [3]*100, 'k-', lw=2, label='Parallel h')
ax.text(5.6, 1, 'g', fontsize=12)
ax.text(5.6, 3, 'h', fontsize=12)

# Transversale (schneidende Gerade)
# Geht durch (1,1) und (3,3) -> Steigung 1
ax.plot([0, 4], [0, 4], 'k--', lw=2, label='Transversale t')

# Winekl oben (an gerade h) ---
# Stufenwinkel Alpha 1 (oben rechts)
ax.add_patch(Arc((3,3), 1,1, angle=0, theta1=0, theta2=45, color='green', lw=2))
ax.text(3.6, 3.2, r'$\alpha_1$', color='green', fontsize=14)

# Scheitelwinkel zu Alpha 1 (unten links) -> für Wechselwinkel relevant
ax.add_patch(Arc((3,3), 1,1, angle=0, theta1=180, theta2=225, color='green', lw=2, ls='--'))
ax.text(2.2, 2.7, r'$\gamma_1$', color='green', fontsize=14)

# --- Winkel unten (an gerade g) ---
# Stufenwinkel Alpha 2 (Korrespondierend zu Alpha 1)
ax.add_patch(Arc((1,1), 1, 1, angle=0, theta1=0, theta2=45, color='green', lw=2))
ax.text(1.7, 1.2, r'$\alpha_2$', color='green', fontsize=14)

# Wechselwinkel (Z-Winkel) zu alpha 1 (unten links an Gerade g)
ax.add_patch(Arc((1,1), 1, 1, angle=0, theta1=180, theta2=225, color='blue', lw=2))
ax.text(0.2, 0.7, r'$\gamma_2$', color='blue', fontsize=14)

# Legende / Erklärung
ax.text(4, 0.5,
        r"Stufenwinkel: $\alpha_1 = \alpha_2$" + "\n" +
        r"Wechselwinkel: $\alpha_1 = \gamma_2$",
        bbox=dict(facecolor='white', edgecolor='black', alpha=0.8), fontsize=12)

plt.show()
```

### 3. Kongruenz und Ähnlichkeit

In der Geometrie vergleichen wir oft Figuren miteinander, dabei gibt es zwei Abstufungen der "Gleichheit":

#### Kongruenz (Deckungsgleichheit)
Zwei Figuren sind **kongruent** ($\cong$), wenn sind in Form und Größe exakt übereinstimmen.
Man könnte eine Figur über die andere legen und nichts würde überstehen.

Matheamtisch bedeutet das: Eine Figur lässt sich durch eine **Kongruenzabbildung** (Verschiebung, Drehung oder Spiegelung) in die andere überführen. Dabei ändern sich weder Seitenlängen noch Winkel.

**Die Kongruenzsätze für Dreicke:**
Um zu bestimmen, ob zwei Dreiecke kongruent sind, müssen nicht alle Eigenschaften des Dreiecks betrachtet werden. Es reicht, wenn folgende Kombinationen übereinstimmen:

1. **SSS:** Alle drei Seiten sind gleich lang.
2. **SWS:** Zwei Seiten und der *eingeschlossene* Winkel sind gleich.
3. **WSW:** Eine Seite und die beiden anliegenden Winkel sind gleich.

#### Ähnlichkeit (Formgleichheit)
Zwei Figuren sind **ähnlich** ($\sim$), wenn sie die gleiche Form haben, aber nicht unbedingt die selbe Größe.
Beispiel: Ein Foto, in das man hinein und hinaus zoomt: Das Bild wird größer, aber die Proportionen bleiben gleich (**zentrische Streckung**).

Mathematisch bedeutet das:

1. Alle entsprechenden **Winkel sind gleich groß**.
2. Alle entstprechenden **Seitenverhältnisse sind konstant** (wird Seite $a$ verdoppelt, müssen auch $b$ und $c$ verdoppelt werden).
3. Die Figur lässt sich *zusätzlich* zu **Kongruenzabbildungen** durch **zentrische Streckung** in eine andere überführen.

```python
# Settings für die Animation
Kongruenz_Aehnlichkeit_Settings = {
    # Settings für die Kongruenz-Animation
    "kongruen_basis_punkte": [(2, 2), (4, 2), (3, 4)],  # Basis-Punkte des Dreiecks für die Kongruenz-Animation
    "kongruenz_animation_drehen": True,             # Ob die Drehung in der Kongruenz-Animation stattfinden soll
    "dreh_winkel_kongruenz": 60,                    # Maximaler Drehwinkel für die Kongruenz-Animation
    "kongruenz_animation_verschieben": True,       # Ob die Verschiebung in der Kongruenz-Animation stattfinden soll
    "verschiebungs_vektor_kongruenz": (4, 3),       # Verschiebungsvektor für die Kongruenz-Animation

    # Settings für die Ähnlichkeits-Animation
    "aehnlichkeits_basis_punkte": [(2, 2), (4, 2), (3, 4)],  # Basis-Punkte des Dreiecks für die Ähnlichkeits-Animation
    "aehnlichkeits_animation_drehen": True,         # Ob die Drehung in der Ähnlichkeits-Animation stattfinden soll
    "dreh_winkel_aehnlichkeit": 30,                 # Maximaler Drehwinkel für die Ähnlichkeits-Animation
    "aehnlichkeits_animation_verschieben": True,   # Ob die Verschiebung in der Ähnlichkeits-Animation stattfinden soll
    "verschiebungs_vektor_aehnlichkeit": (4, 1),    # Verschiebungsvektor für die Ähnlichkeits-Animation
    "aehnlichkeit_animation_skalieren": True,      # Ob die Skalierung in der Ähnlichkeits-Animation stattfinden soll
    "skalierungs_faktor_aehnlichkeit": 2.5,         # Skalierungsfaktor für die Ähnlichkeits-Animation

    # Allgemeine Animationseinstellungen
    "figsize": (12, 6),
    "ax1_xlim": (0, 10),
    "ax1_ylim": (0, 10),
    "ax2_xlim": (0, 10),
    "ax2_ylim": (0, 15),
    "interval": 100,
    "frames": 100
}
```

```python
# ANIMATION: Kongruenz vs Ähnlichkeit

from matplotlib.animation import FuncAnimation
from IPython.display import HTML

cfg = Kongruenz_Aehnlichkeit_Settings

# Setup der Figur
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=cfg["figsize"])
fig.suptitle("Vergleich: Kongruenz vs Ähnlichkeit", fontsize=16)

# --- Linker Plot: Kongruenz (Verschiebung + Drehung) ---
ax1.set_xlim(cfg["ax1_xlim"])
ax1.set_ylim(cfg["ax1_ylim"])
ax1.set_aspect('equal')
ax1.grid(True, alpha=0.3)
ax1.set_title("Kongruenz: Verschiebung + Drehung", fontsize=14)


# Start-Dreieck
triangle_cong_x = np.array([x for x, y in cfg["kongruen_basis_punkte"]] + [cfg["kongruen_basis_punkte"][0][0]])
triangle_cong_y = np.array([y for x, y in cfg["kongruen_basis_punkte"]] + [cfg["kongruen_basis_punkte"][0][1]])
triangle_cong, = ax1.plot(triangle_cong_x, triangle_cong_y, 'b-', lw=2, label='Original')
triangle_cong_shadow, = ax1.plot(triangle_cong_x, triangle_cong_y, 'b--', alpha=0.3) # Schatten der Startposition

# --- Rechter Plot: Ähnlichkeit (Skalierung + Drehung) ---
ax2.set_xlim(cfg["ax2_xlim"])
ax2.set_ylim(cfg["ax2_ylim"])
ax2.set_aspect('equal')
ax2.grid(True, alpha=0.3)
ax2.set_title("Ähnlichkeit: Skalierung + Verschiebung + Drehung", fontsize=14)

# Start-Dreieck
triangle_sim_x = np.array([x for x, y in cfg["aehnlichkeits_basis_punkte"]] + [cfg["aehnlichkeits_basis_punkte"][0][0]])
triangle_sim_y = np.array([y for x, y in cfg["aehnlichkeits_basis_punkte"]] + [cfg["aehnlichkeits_basis_punkte"][0][1]])
triangle_sim, = ax2.plot(triangle_sim_x, triangle_sim_y, 'r-', lw=2, label='Original')
triangle_sim_shadow, = ax2.plot(triangle_sim_x, triangle_sim_y, 'r--', alpha=0.3) # Schatten der Startposition




def init():
    # Schatten zeichnen
    triangle_cong_shadow.set_data(triangle_cong_x, triangle_cong_y)
    triangle_sim_shadow.set_data(triangle_sim_x, triangle_sim_y)

    # Text-Anmerkung: Bei dieser Animation bleiben die Formen kongruent/ähnlich,
    # nur ihre Position, Orientierung und Größe ändert sich. = Unter beiden Plots
    fig.text(0.5, 0.02, 'Die Formen bleiben kongruent/ähnlich - nur Position, Orientierung und Größe ändern sich', 
             ha='center', fontsize=12, style='italic')

    return triangle_cong, triangle_sim

def update(frame):
    # --- Kongruenz: Verschiebung und leichte Drehung ---
    # Frame 0-100
    progress = frame / 100

    if cfg["kongruenz_animation_verschieben"]:
        # Verschiebung (Translation) um (4,3)
        shift_x = cfg["verschiebungs_vektor_kongruenz"][0] * progress
        shift_y = cfg["verschiebungs_vektor_kongruenz"][1] * progress

        current_cong_x = triangle_cong_x + shift_x
        current_cong_y = triangle_cong_y + shift_y
        
    else:
        current_cong_x = triangle_cong_x
        current_cong_y = triangle_cong_y
        
    if (cfg["kongruenz_animation_drehen"]):
        # Leichte Drehung
        angle_cong = np.radians(cfg["dreh_winkel_kongruenz"] * progress)
        rot_matrix_cong = np.array([[np.cos(angle_cong), -np.sin(angle_cong)],
                                    [np.sin(angle_cong),  np.cos(angle_cong)]])
        rotated_cong = rot_matrix_cong @ np.vstack((current_cong_x - (shift_x + 2), current_cong_y - (shift_y + 2)))
        current_cong_x = rotated_cong[0, :] + (shift_x + 2)
        current_cong_y = rotated_cong[1, :] + (shift_y + 2)
    
    triangle_cong.set_data(current_cong_x, current_cong_y)

    # --- Ähnlichkeit: Skalierung und Drehung ---
    # Skalierungsfakotr von 1.0 bis konfiguriertem Faktor

    if (cfg["aehnlichkeit_animation_skalieren"]):
        scale_factor = 1.0 + (cfg["skalierungs_faktor_aehnlichkeit"] - 1.0) * progress

        current_sim_x = (triangle_sim_x) * scale_factor
        current_sim_y = (triangle_sim_y) * scale_factor

    else:
        scale_factor = 1.0
        current_sim_x = triangle_sim_x
        current_sim_y = triangle_sim_y

    if cfg["aehnlichkeits_animation_verschieben"]:
        # Verschiebung (Translation) um (4,3)
        shift_x_sim = cfg["verschiebungs_vektor_aehnlichkeit"][0] * progress
        shift_y_sim = cfg["verschiebungs_vektor_aehnlichkeit"][1] * progress

        current_sim_x = current_sim_x + shift_x_sim
        current_sim_y = current_sim_y + shift_y_sim
    
    if (cfg["aehnlichkeits_animation_drehen"]):
        angle_sim = np.radians(60 * progress) # bis zu 60
        rot_matrix_sim = np.array([[np.cos(angle_sim), -np.sin(angle_sim)],
                                [np.sin(angle_sim), np.cos(angle_sim)]])
        rotated_sim = rot_matrix_sim @ np.vstack((current_sim_x - (scale_factor * 2.0), current_sim_y - (scale_factor * 2.5))) 
        current_sim_x = rotated_sim[0, :] + (scale_factor * 2.0)
        current_sim_y = rotated_sim[1, :] + (scale_factor * 2.5)

    triangle_sim.set_data(current_sim_x, current_sim_y)

    return triangle_cong, triangle_sim

# Animation erstellen
anim = FuncAnimation(fig, update, frames=101, init_func=init, blit=True, interval=30)

plt.close()
HTML(anim.to_jshtml())
```

### Verständnis-Check

**Aufgabe 1: Winkeljagd**
Zwei Geraden schneiden sich. Einer der entstehenden Winkel beträgt $\alpha = 35°$.
Berechne die anderen drei Winkel $\beta, \gamma, \delta$ unter Verwendung der Begriffe *Scheitelwinkel* und *Nebenwinkel*.

**Aufgabe 2: Parallelen-Check**
Eine Gerade schneidet zwei andere Geraden.
- An der oberen Kreuzung ist der Winkel oben rechts $50°$.
- An der unteren Kreuzung ist der Winkel unten links $55°$.
Sind die beiden Geraden parallel? Begründe deine Antwort mit *Scheitel-* und *Stufenwinkeln*.

**Aufgabe 3: Ähnlichkeit**
Du hast ein Foto von einem Dreieck mit den Seitenlängen $a=3$ cm, $b=4$ cm, $c=5$ cm.
Du vergrößerst das Foto auf dem Kopierer um den Faktor 2 (Skalierung).
- Wie lang sind die Seiten des neuen Dreiecks?
- Verändern sich die Innenwinkel des Dreiecks?
- Sind das ursprüngliche und das kopierte Dreieck kongruent?

### Lösungen

**Lösung 1:**
- $\gamma = 35°$ (Scheitelwinkel zu $\alpha$, daher gleich groß).
- $\beta = 180° - 35° = 145°$ (Nebenwinkel zu $\alpha$).
- $\delta = 145°$ (Scheitelwinkel zu $\beta$).

**Lösung 2:**
Nein, sie sind nicht parallel.
- Der Winkel oben rechts ist $50°$. Sein Scheitelwinkel (unten links an der oberen Kreuzung) ist ebenfalls $50°$.
- Wären die Geraden parallel, müsste dieser Winkel (50°) als *Stufenwinkel* zum Winkel unten links an der unteren Kreuzung (55°) identisch sein.
- Da $50° \neq 55°$, sind die Geraden nicht parallel.

**Lösung 3:**
- Die neuen Seitenlängen sind $a'=6$ cm, $b'=8$ cm, $c'=10$ cm (jede Seite $\times 2$).
- Die Innenwinkel verändern sich **nicht**. Bei Ähnlichkeit bleiben Winkel erhalten.
- Nein, sie sind **nicht kongruent**, da sie nicht deckungsgleich sind (unterschiedliche Größe). Sie sind **ähnlich**.

### Glossar (Deutsch - Englisch)

- **Schnittpunkt:** Intersection (point)
- **Nebenwinkel:** Supplementary angles (specifically adjacent angles on a line)
- **Scheitelwinkel:** Vertical angles
- **Stufenwinkel:** Corresponding angles
- **Wechselwinkel:** Alternate angles
- **Parallel:** Parallel
- **Transversale:** Transversal
- **Kongruenz:** Congruence
- **Ähnlichkeit:** Similarity
- **Deckungsgleich:** Superimposable / Congruent
- **Verschiebung:** Translation
- **Drehung:** Rotation
- **Spiegelung:** Reflection
- **Skalierung / Zentrische Streckung:** Scaling / Homothety / Dilation
