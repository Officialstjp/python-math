# Vektoren

Viele physikalische Größen lassen sich nicht allein durch ihren Betrag vollständig charakterisieren. Zusätzlich ist noch die Angabe einer Richtung notwendig.
Solche gerichtete Größen bezeichnet man als Vektoren.








## Definition

Ein Vektor entspricht der Menge aller Pfeile mit gleicher Richutng, gleicher Orientierung und gleicher Länge.

Ein einzelner Pfeil dieser Menge ist ein Repräsentant (Stellvertreter) des entsprechenden Vektors.


Die Länge eines Vektors wird entweder in Betragsnotation oder Normnotation angegeben: $| \vec{A} |$


Der **Nullvektor** beschreibt den Vektor bei Punkt (0,0)

Der **Einheitsvektor** ist ein Vektor, welcher die selbe Richtung wie ein anderer gegebener Vektor hat und die Länge $1$ hat.

Der **Einheitsvektor** ist definiert:
$$
\vec{a_0} = \frac{\vec{a}}{|\vec{a}|} = \frac{1}{|\vec{a}|} \cdot \vec{a} 
$$

## Spitze - Schaft

Möchte man den Vektor zwischen zwei Punkt-Vektoren berechnen, nimmt man die Koordinaten des Spitzvektors und subtrahiert diese mit den Koordinaten des Schaftvektors.

# Regeln für Vektoren

- Der Vektor $\vec{BA} = -\vec{AB}$ heißt **Gegenvektor** des Vektors $\vec{a} = \vec{AB}$.
Der Gegenvektor $-\vec{a}$ hat gleiche Länge und gleiche Richtung, aber entgegengesetzte Orientierung zum Vektor $\vec{a}$.

- Vektoren werden komponentenweise **addiert** bzw. **subtrahiert**.
$$
\vec{a} \pm \vec{b} = \begin{pmatrix} a_x \\ a_y \end{pmatrix} \pm \begin{pmatrix} b_x \\ b_y \end{pmatrix}
= \begin{pmatrix} a_x \pm b_x \\ a_y \pm b_y \end{pmatrix}
$$

Eine Vektoraddition wird grafisch durchgeführt, indem an die Spitze des ersten Vektores $\vec{a}$ der Schaft (Anfangspunkt) des zweiten Vektors $\vec{b}$ gesetzt wird.

Die Vektorsumme $\vec{a} + \vec{b}$ ergibt sich durch den Pfeil, der a Schaft des ersten Vektors beginnt und an der Spitze des zweiten Vektors endet.

- Vektoren werden komponentenweise mit einer reellen Zahl (einem Skalar) multiliziert.
$$
c = \vec{a} = c \cdot \begin{pmatrix} a_x \\ a_y \end{pmatrix} = \begin{pmatrix} c \cdot a_x \\ c \cdot a_y \end{pmatrix}
$$

## Rechenregeln
Vektoren dürfen addiert, subtrahiert und mit einem Skalar multipliziert werden.

Bei der Skalarmultiplikation werden die Vektoren zeilenweise (x, y) mit den zugehörigen Skalar (Faktor-Zahl mit dem multipliziert wird) multiplizert. 



## Audrückung in Kräften

Vektoren werden häufig verwendet um Physikalische Kräfte zu beschreiben.

Zwei Vektoren in einem Koordinatensystem können als "Vektor-Parallelogramm" dargestellt werden. Die Mittellinie (jene, die vom Anfangspunkt zum gegenüberliegenden Ende des Parallelogramms existiert), stellt den resultierenden Vektor dar.
