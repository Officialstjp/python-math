## Geometrische Grundbegriffe

Die Geometrie baut auf wenigen, intuitiven Elementen auf:

- **Punkt:** Eine exakte Position in einem Raum ohne Ausdehnung. Er ist die Grundlage für alle geometrischen Figuren, bezeichnet meist mit Großbuchstaben (z.B. $P, A, B$).
- **Gerade:** Eine unendlich lange, gerade ohne Anfang und Ende.
- **Strahl (oder Halbgerade):** Beginnt an einem Anfangspunkt, und verläuft unendlich in eine Richtung.
- **Strecke:** Die kürzeste Verbindung zwischen zwei Punkten $A und B$. Sie hat eine messbare Länge.

Ein Winkel wird generell in **Grad** gemessen, wobei eine volle Umdrehung $360°$ entspricht. Die Größe des Winkels beschreibt die "Öffnung" zwischen den beiden Strahlen.

```python
"""Geometrische Grundbegriffe zeichnen"""
import matplotlib.pyplot as plt
import numpy as np

# Setup des Plots
fig, ax = plt.subplots(figsize=(10, 4))
ax.set_xlim(-1, 13)
ax.set_ylim(0, 4)
ax.axis('off') # Achsen ausblenden für rein geometrische Darstellung

# 1. Punkt
ax.plot(1, 3, 'ro') # ro = roter Punkt
ax.text(1, 3.2, '$P$', fontsize=14, ha='center')
ax.text(1, 1, 'Punkt', fontsize=12, ha='center', fontweight='bold')

# 2. Gerade (angedeutet durch Pfeile an beiden Enden oder Überhang)
ax.plot([3, 5], [3, 3], 'b-', linewidth=2)
# Pfeile simulieren Unendlichkeit
ax.annotate('', xy=(2.8, 3), xytext=(3, 3), arrowprops=dict(arrowstyle='->', color='blue'))
ax.annotate('', xy=(5.2, 3), xytext=(5, 3), arrowprops=dict(arrowstyle='->', color='blue'))
ax.text(4, 1, 'Gerade', fontsize=12, ha='center', fontweight='bold')

# 3. Strahl (Ein fester Punkt, ein offenes Ende)
ax.plot(7, 3, 'go') # Startpunkt
ax.plot([7, 9], [3, 3], 'g-', linewidth=2)
ax.annotate('', xy=(9.2, 3), xytext=(9, 3), arrowprops=dict(arrowstyle='->', color='green'))
ax.text(8, 1, 'Strahl', fontsize=12, ha='center', fontweight='bold')

# 4. Strecke (Zwei feste Punkte)
ax.plot([10, 12], [3, 3], 'k-', linewidth=2, marker='o') # marker='o' macht Punkte an Enden
ax.text(10, 3.2, '$A$', fontsize=12)
ax.text(12, 3.2, '$B$', fontsize=12)
ax.text(11, 1, 'Strecke $\\overline{AB}$', fontsize=12, ha='center', fontweight='bold')

plt.title("Geometrische Grundelemente")
plt.tight_layout()
plt.show()
```

## Der Winkel

**Definition**
Ein **Winkel** entsteht, wenn zwei Strahlen (die sogennanten **Schenkel**) von einem gemeinsamen Anfangspunkt (dem Scheitel) ausgehen.

**Notation**
Winkel werden meist mit griechischen Kleinbuchstaben ($\alpha, \beta, \gamma,...$) bezeichnet. Alternativ können auch die Punkten auf den Schenkeln verwendet werden:
$\angle ABC$ bedeutet, der Scheitel ist bei B, und der Winkel wird von $A$ über $B$ nach $C$ aufgespannt (meist gegen den Uhrzeigersinn).

**Maßeinheiten: Grad vs Bogenmaß**
1. **Gradmaß**: Ein Vollkreis wird in $360°$ unterteilt.
2. **Bogenmaß**: Ein Winkel wird durch die Länge eines Kreisbogens am Einheitskreis ($r = 1$) definiert. Ein Vollkreis entspricht dem Umfang $2\pi$

**Umrechnung:**
$$
\alpha_{\text{rad}} = \alpha_{\text{deg}} \cdot \frac{\pi}{180°}
$$
$$
\alpha_{\text{deg}} = \alpha_{\text{rad}} \cdot \frac{180°}{\pi}
$$

Wichtige Werte:
- $90° = \frac{\pi}{2}$
- $180° = \pi$
- $360° = 2\pi$

```python
"""Aufbau eines Winkel zeichnen"""
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Arc

# --- Plot Setup ---
fig, ax = plt.subplots(figsize=(8,6))
ax.set_aspect('equal')
ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-0.2, 1.2)
ax.set_title("Aufbau eines Winkels", fontsize=16)
ax.axis('off') # Achsen ausblenden

# --- Elemente definieren ---
scheitelpunkt = (0, 0)
strahl1_ende = (1, 0)
winkel_grad = 45
winkel_rad = np.deg2rad(winkel_grad) # In Numpy müssen Grad zuerst in Radianten konvertiert werden, um sin() und cos() zu berechnen.
strahl2_ende = (np.cos(winkel_rad), np.sin(winkel_rad))

# --- Elemente zeichnen ---
# Scheitelpunkt
ax.plot(scheitelpunkt[0], scheitelpunkt[1], 'ko', markersize=8, label="Scheitelpunkt")

# Strahl 1 (Schenkel 1)
ax.arrow(scheitelpunkt[0], scheitelpunkt[1], strahl1_ende[0], strahl1_ende[1], head_width=0.04,
         head_length=0.08, fc='blue', ec='blue', length_includes_head=True)
ax.text(0.6, -0.1, "Schenkel 1 (Strahl)", color='blue')

# Strahl 2 (Schenkel 2)
ax.arrow(scheitelpunkt[0], scheitelpunkt[1], strahl2_ende[0], strahl2_ende[1],
         head_width=0.04, head_length=0.08, fc='red', ec='red')
ax.text(0.5, 0.5, "Schenkel 2 (Strahl)", color='red')

# Winkelbogen
winkel_bogen = Arc(scheitelpunkt, 0.5, 0.5, angle=0, theta1=0, theta2=winkel_grad,
                   color='green', lw=2, label="Winkel")
ax.add_patch(winkel_bogen)
ax.text(0.35, 0.15, r"$\alpha$", color='green', fontsize=20)

# --- Legende ---
# Manuelle Anordnung der Texte als Legende
ax.text(-1.1, 1.0, "Scheitelpunkt", fontsize=12, va='center')
ax.plot(-1.15, 1.0, 'ko', markersize=8)
ax.text(-1.1, 0.9, "Schenkel (Strahlen)", fontsize=12, va='center')
ax.plot([-1.2, -1.1], [0.85, 0.85], color='black', lw=2)
ax.text(-1.1, 0.8, r"Winkel ($\alpha$)", fontsize=12, va='center')
ax.add_patch(Arc((-1.15, 0.75), 0.1, 0.1, theta1=0, theta2=60, color='green', lw=2))

plt.show()
```

## Winkelarten

Je nach ihrer Größe in Grad lassen sich Winkel in verschiedene Kategorien einteilen:

- **Nullwinkel:** $\alpha = 0°$
  Beide Schenkel liegen aufeinander (auf der selben Linie), es gibt keine Öffnung.
  
- **Spitzer Winkel:** $0° < \alpha < 90°$
  Der Winkel ist kleiner als ein rechter Winkel ($<90°$).

  
- **Rechter Winkel:** $\alpha = 90°$
  Die Schenkel stehen **senkrecht** (orthogonal) aufeinander.
  
- **Stumpfer Winkel:** $90° < \alpha < 180°$
  Der Winkel ist größer als ein rechter Winkel ($<90°$ und kleiner als ein gestreckter Winkel ($<180°$).
  
- **Gestreckter Winkel:** $\alpha = 180°$
  Die beiden Schenkel bilden eine gerade Linie.
  
- **Überstumpfer Winkel (erhabener Winkel):** $180° < \alpha < 360°$
  Ein Winkel, der größer als ein gestrecker Winkel ist.

- **Vollwinkel:** $\alpha = 360°$
  Der Winkel beschreibt eine vollständige Umdrehung. Wie beim Nullwinkel liegt der zweite Schenkel wieder auf dem ersten.

```python
"""Verschieden Winkelarten zeichnen"""
def plot_angle(ax, angle_deg, title):
    """Hilfsfunktion zum Zeichnen eines Winkels auf einer gegebenen Achse."""
    ax.set_aspect('equal')
    ax.set_xlim(-1.1, 1.1)
    ax.set_ylim(-0.6, 1.2)
    ax.axis('off')
    ax.set_title(title, fontsize=12)

    Scheitelpunkt = (0, 0)

    # Erster Schenkel immer nach rechts
    ax.arrow(0, 0, 1, 0, head_width=0.07, head_length=0.1, fc='gray', ec='gray', length_includes_head=True)

    # Zweiter Schenkel basierend auf dem Winkel
    if angle_deg > 0:
        angle_rad = np.deg2rad(angle_deg)
        end_x = np.cos(angle_rad)
        end_y = np.sin(angle_rad)
        ax.arrow(0, 0, end_x, end_y, head_width=0.07, head_length=0.1, fc='black', ec='black', length_includes_head=True)

    # Winkelbogen zeichen
    if 0 < angle_deg < 360:
        bogen = Arc(scheitelpunkt, 0.6, 0.6, angle=0, theta1=0, theta2=angle_deg, color='blue', lw=1.5)
        ax.add_patch(bogen)

    # Spezialsymbol für rechten Winkel
    if angle_deg == 90:
        ax.plot([0.2, 0.2, 0], [0, 0.2, 0.2], color='blue', lw=1.5)

    # Text für Winkelgröße
    if 0 < angle_deg < 360:
        label_rad = np.deg2rad(angle_deg / 2)
        ax.text(0.4 * np.cos(label_rad), 0.4 * np.sin(label_rad), f"{angle_deg}°",
                ha='center', va='center', fontsize=10, color='darkblue')

# --- Plot erstellen ---
fig, axes = plt.subplots(2, 3, figsize=(15, 9))
fig.suptitle("Übersicht der Winkelarten", fontsize=18)

# Einzelne Winkel plotten
plot_angle(axes[0, 0], 45, "Spitzer Winkel (0° < α < 90°)")
plot_angle(axes[0, 1], 90, "Rechter Winkel (α = 90°)")
plot_angle(axes[0, 2], 135, "Stumpfer Winkel (90° < α < 180°)")
plot_angle(axes[1, 0], 180, "Gestreckter Winkel (α = 180°)")
plot_angle(axes[1, 1], 240, "Überstumpfer Winkel (180° < α < 360°)")
plot_angle(axes[1, 2], 360, "Vollwinkel (α = 360°)")

# Layout anpassen und anzeigen
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

```python
"""Animation der Winkelarten"""
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Funktion zur Bestimmung der Winkelart
def get_angle_type(deg):
    if deg == 0: return "Nullwinkel"
    if 0 < deg < 90: return "Spitzer Winkel"
    if deg == 90: return "Rechter Winkel"
    if 90 < deg < 180: return "Stumpfer Winkel"
    if deg == 180: return "Gestreckter Winkel"
    if 180 < deg < 360: return "Überstumpfer Winkel"
    if deg == 360: return "Vollwinkel"
    return ""

# Setup
fig, ax = plt.subplots(figsize=(6,6))
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 1.5)
ax.set_aspect('equal')
ax.axis('off')

# Statische Elemente
scheitel = (0,0)
ax.plot(0,0,'ko') # Scheitelpunkt
ax.plot([0,1.2],[0,0], 'k-', lw=2) # Fester Schenkel (x-Achse)
ax.text(1.3, 0, 'Start', ha='left', va='center')

# Dynamische Elemente
moving_arm, = ax.plot([], [], 'r-', lw=2) # Der bewegliche Schenkel
arc_fill, = ax.fill([], [], 'r', alpha=0.2) # Füllung des Winkels
text_label = ax.text(0, -1.2, '', ha='center', fontsize=14, fontweight='bold', color='darkred')
degree_label = ax.text(0, 1.2, '', ha='center', fontsize=12)

def init():
    moving_arm.set_data([], [])
    return moving_arm, text_label

def update(frame):
    deg = 0
    # Wir geben jedem Grad einen Frame, und fügen Pausen für spezielle Winkelarten hinzu
    # Frames 0-30 = Nullwinkel
    if frame < 31:
        # Nullwinkel, nichts zu zeichnen
        deg = 0

    # Frames ab 31-120 (90+30 Pause) = normaler spitzer Winkel
    if 31 <= frame < 120:
        deg = frame - 30

    # Frames 120 - 150 (90+60 Pause) = rechter Winkel
    if 120 <= frame < 150:
        deg = 90

    # Frames 151 - 240 = normaler stumpfer Winkel
    if 150 <= frame < 240:
        deg = frame - 60

    # Frame 240 - 270 (180+90 Pause) = gestreckter Winkel
    if 240 <= frame < 270:
        deg = 180

    # Frames 270 - 450 (360+90 Pause) = überstumpfer Winkel
    if 270 <= frame < 450:
        deg = frame - 90

    # Frames 450 - 480 = Vollwinkel
    if 450 <= frame < 481:
        deg = 360
        circle = plt.Circle((0,0), 0.5, color='r', alpha=0.2)
        ax.add_patch(circle)

    # Radianten berechnen für numpy trig Funktionen
    angle_rad = np.radians(deg)

    # Koordinaten des beweglichen Schenkels
    x = 1.2 * np.cos(angle_rad)
    y = 1.2 * np.sin(angle_rad)
    moving_arm.set_data([0, x], [0, y])

    # Winkelbogen zeichnen (Polygon füllen)
    arc_angles = np.linspace(0, angle_rad, num=50)
    arc_x = np.concatenate(([0], 0.5 * np.cos(arc_angles), [0]))
    arc_y = np.concatenate(([0], 0.5 * np.sin(arc_angles), [0]))

    arc_fill.set_xy(np.column_stack((arc_x, arc_y)))

    # Text Updates
    degree_label.set_text(f"{deg}°")
    text_label.set_text(get_angle_type(deg))

    return moving_arm, text_label, degree_label

# Animation erstellen
# (von 0 bis 360 in 2-Grad-Schritten + 120 Frames aus
# 4x 30 Frames um Nullwinkel, rechter Winkel, Gestreckter Winkel und Vollwinkel zu zeigen)
frames = np.arange(0, 481, 2)
ani = FuncAnimation(fig, update, frames=frames, init_func=init, blit=True, interval=30)

# Um die Animation anzuzeigen:
plt.close()
HTML(ani.to_jshtml())
```

## Verständnis-Check

Versuch folgende Aufgaben zu beantworten (Lösungen in nächster Zelle):

**Aufgabe 1:**
Welche Art von Winkel ist ein Winkel mit der Größe 270°?

**Aufgabe 2:**
Ein Uhrzeiger bewegt sich von der 12-Uhr-Position zur 4-Uhr-Position. 
Welchen Winkel hat er überstrichen und um welche Winkelart handelt es sich?

**Aufgabe 3:**
Zwei Winkel, $\alpha=40°$ und $\beta=50°$, werden aneinandergelegt, sodass sie einen gemeinsamen Schenkel haben. Welchen Winkel bilden ihre äußeren Schenkel und um welche Art handelt es sich?

### Lösungen

<details><summary>Lösung Aufgabe 1</summary>

Ein Winkel von 270° ist größer als 180° und kleiner als 360°. Daher handelt es sich um einen **überstumpfen (oder erhabenen) Winkel**.

</details>

<details><summary>Lösung Aufgabe 2</summary>

Das Ziffernblatt einer Uhr ist in 12 Stunden unterteilt, was einer vollen Umdrehung von 360° entspricht. Die Bewegung von einer Stundenziffer zur nächsten ist also $360° / 12 = 30°$.
Wenn der Zeiger von 12 Uhr nach 4 Uhr wandert, überstreicht er 4 solcher Abschnitte.
Der Gesamtwinkel beträgt also $4 \times 30° = 120°$.
Da $90° < 120° < 180°$, ist dies ein **stumpfer Winkel**.

</details>

<details><summary>Lösung Aufgabe 3</summary>

Die Größe des neuen Winkels ist die Summe der beiden Einzelwinkel: $\gamma = \alpha + \beta = 40° + 50° = 90°$.
Ein Winkel von 90° ist ein **rechter Winkel**. Man nennt solche Winkel, die sich zu 90° ergänzen, auch **Komplementärwinkel** - ein Begriff, den wir später noch genauer betrachten werden.

</details>
