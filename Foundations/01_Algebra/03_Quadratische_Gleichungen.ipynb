## Quadratische Funktionen - Definition und Grundlagen

### Definition
Eine **quadratische Funktion** ist eine Funktion der Form:
$$f(x) = ax^2 + bx + c$$
wobei $a, b, c \in \mathbb{R}$ und **$a \neq 0$** (sonst wäre es linear).

### Die drei Koeffizientenrollen

| Koeffizient | Bedeutung | Einfluss |
|---|---|---|
| **$a$** (Leitkoeffizient) | Öffnungsrichtung & Streckung | $a > 0$ → oben offen; $\|a\|$ → Breite der Parabel |
| **$b$** | Asymmetrie & Achsenverschiebung | Beeinflusst die Position des Scheitels |
| **$c$** | y-Achsenabschnitt | $f(0) = c$ |



```python
import numpy as np
import matplotlib.pyplot as plt

# --- Quadratische Funktion  ---
# Merken: Höchster Punkt - "Scheitel"

def quadratic_function(a, b, c, x):
    return a*x**2 + b*x + c

# --- Plot definieren ---
plt.figure(figsize=(10, 6))
plt.grid(True)
plt.axhline(0, color='black', linewidth=1.5) # x-achse
plt.axvline(0, color='black', linewidth=1.5) # y-achse
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.title("Quadratic Function: f(x) = ax² + bx +c")

# --- x-Werte generieren
x_values = np.linspace(-10, 10, 400) # 400 Punkte von -10 bis 10

# ---- Functionen zeichnen
# 1. Normalfunktion: y = x² (a=1, b=0, c=0)
y1 = quadratic_function(1, 0, 0, x_values)
plt.plot(x_values, y1, label="a=1, b=0, c=0 (y = x²)")

# 2. Einflss von 'a': Funktion offen nach unten
y2 = quadratic_function(-1, 0, 0, x_values)
plt.plot(x_values, y2, label="a=-1, b=0, c=0 (y = -x²)")

# 3. Einfluss von 'b': Verschiebt die Parabel auf der x-achse
y3 = quadratic_function(1, 4, 0, x_values)
plt.plot(x_values, y3, label="a=1, b=0, c=4 (y = x² + 4)")

# 4. Einfluss von 'c': Verschiebt die Parabel auf der y-achse
y4 = quadratic_function(1, 0, -4, x_values)
plt.plot(x_values, y4, label="a=1, b=-4, c=0 (y = x² - 4x)")

plt.legend()
plt.ylim(-10, 20) # Anzeige-limits für y-achse
plt.show()

```

### Normalform vs. Scheitelpunktform vs. Faktorisierte Form

Die gleiche Funktion kann in drei Formen geschrieben werden:

1. **Normalform**: $f(x) = ax^2 + bx + c$
2. **Scheitelpunktform**: $f(x) = a(x - h)^2 + k$ wobei $(h, k)$ der Scheitelpunkt ist
3. **Faktorisierte Form**: $f(x) = a(x - r_1)(x - r_2)$ wobei $r_1, r_2$ die Nullstellen sind

Der Scheitelpunkt liegt bei:
$$x_s = -\frac{b}{2a}, \quad y_s = f(x_s) = c - \frac{b^2}{4a}$$

```python
import numpy as np
import matplotlib.pyplot as plt

# Eine Funktion in drei verschiedenen Formen
# Beispiel: f(x) = x² - 4x + 3

a, b, c = 1, -4, 3

# Scheitelpunkt berechnen
x_s = -b / (2*a)
y_s = c - (b**2) / (4*a)

print(f"Normalform: f(x) = {a}x² + ({b})x + {c}")
print(f"Scheitelpunkt: ({x_s}, {y_s})")
print(f"Scheitelpunktform: f(x) = {a}(x - {x_s})² + {y_s}")

# Nullstellen (Faktorisierte Form)
discriminant = b**2 - 4*a*c
r1 = (-b - np.sqrt(discriminant)) / (2*a)
r2 = (-b + np.sqrt(discriminant)) / (2*a)
print(f"Faktorisierte Form: f(x) = {a}(x - {r1})(x - {r2})")

# --- Visualisierung ---
x = np.linspace(-1, 5, 300)
y = a*x**2 + b*x + c

fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(x, y, 'b-', linewidth=2, label='f(x) = x² - 4x + 3')
ax.axhline(0, color='k', linewidth=0.5)
ax.axvline(0, color='k', linewidth=0.5)

# Scheitelpunkt markieren
ax.plot(x_s, y_s, 'ro', markersize=10, label=f'Scheitelpunkt ({x_s}, {y_s})')

# Nullstellen markieren
ax.plot([r1, r2], [0, 0], 'go', markersize=8, label=f'Nullstellen: x₁={r1}, x₂={r2}')

# y-Achsenabschnitt markieren
ax.plot(0, c, 'mo', markersize=8, label=f'y-Achsenabschnitt: (0, {c})')

# Symmetrieachse einzeichnen
ax.axvline(x_s, color='r', linestyle='--', alpha=0.5, label=f'Symmetrieachse: x = {x_s}')

ax.set_xlabel('x')
ax.set_ylabel('f(x)')
ax.set_title('Quadratische Funktion: Die drei Formen')
ax.grid(True, alpha=0.3)
ax.legend()
ax.set_ylim(-2, 8)
plt.tight_layout()
plt.show()
```

```python
import numpy as np

def find_roots(a, b, c):
    """
    Berechnet die Nullstellen einer quadratischen Funktion ax² + bx + c = 0.

    Args:
        a (float): Die Koeffiziente von x²
        b (float): Die Koeffiziente von x.
        c (float): Konstante (y-shift)

    Returns:
        Tuple mit den Nullstellen. Die Anzahl der Nullstellen aus [0,2]
    """
    print(f"Solving for: {a}x² + {b}x + {c} = 0")

    # Schritt 1: Falls es keine quadratischen Funktion ist
    if a == 0:
        print("This is not a quadratic equation.")
        if b != 0:
            return (-c / b,) # Lineare Funktion
        else:
            return() # Keine Lösung oder unendliche Lösung

    # Schritt 2: Diskriminante berechnen
    discriminant = (b**2) - (4*a*c)
    print(f"The discriminant is: {discriminant}")

    # Schritt 3: Basierend auf die Diskriminante, Anzahl der Nullstellen berechnen
    if discriminant > 0:
        print("There are two distinct real roots (Nullstellen).")
        x1 = (-b - np.sqrt(discriminant)) / (2*a)
        x2 = (-b + np.sqrt(discriminant)) / (2*a)
        return (x1, x2)
    elif discriminant == 0:
        print("There is one real root (Nullstelle).")
        x = -b / (2*a)
        return (x,)
    else: # discriminant < 0
        print("There are no real roots. (Nullstellen).")
        # In höherer Mathematik, würden hier komplexe/imaginäre Zahlen in Frage kommen
        # Hier, return empty tuple
        return ()

# --- Testen ---
y1 = [1,0,0]
y2 = [-1,0,0]
y3 = [1,0,4]
y4 = [1,-4,0]

for y in (y1,y2,y3,y4):
    a, b, c = y[0], y[1], y[2]
    roots = find_roots (y[0],y[1],y[2])
    print(f"\nThe roots are: {roots}")
    print("\n" + "-"*20 + "\n")
```

## Die Diskriminante

Für die quadratische Funktion $f(x) = ax^2 + bx + c$ definieren wir die **Diskriminante**:
$$\Delta = b^2 - 4ac$$


Die Diskriminante bestimmt die Anzahl und Art der Nullstellen:

### Drei Fälle

| Fall | Diskriminante | Nullstellen | Geometrische Bedeutung |
|---|---|---|---|
| **Fall 1** | $\Delta > 0$ | Zwei verschiedene reelle Nullstellen | Parabel schneidet die x-Achse zweimal |
| **Fall 2** | $\Delta = 0$ | Eine (doppelte) Nullstelle | Parabel berührt die x-Achse (Scheitelpunkt liegt auf x-Achse) |
| **Fall 3** | $\Delta < 0$ | Keine reellen Nullstellen | Parabel hat keinen Kontakt mit x-Achse |

**Hinweis**: Falls $\Delta < 0$, existieren komplexe Lösungen der Form $x = \frac{-b \pm i\sqrt{\|\Delta\|}}{2a}$, die an der Universität behandelt werden.

### Die Lösungsformel (ABC-Formel / Mitternachtsformel)
$$x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a} = \frac{-b \pm \sqrt{\Delta}}{2a}$$

```python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
x = np.linspace(-4, 4, 300)

# Fall 1: Delta > 0 (zwei Nullstellen)
a1, b1, c1 = 1, 0, -2
y1 = a1*x**2 + b1*x + c1
delta1 = b1**2 - 4*a1*c1
roots1 = [(-b1 + np.sqrt(delta1))/(2*a1), (-b1 - np.sqrt(delta1))/(2*a1)]

axes[0].plot(x, y1, 'b-', linewidth=2)
axes[0].axhline(0, color='k', linewidth=0.5)
axes[0].axvline(0, color='k', linewidth=0.5)
axes[0].plot(roots1,[0,0], 'ro', markersize=8)
axes[0].grid(True, alpha=0.3)
axes[0].set_title(f'$\Delta$ = {delta1} = > 0\n(Zwei Nullstellen)', fontsize=14)
axes[0].set_ylim(-3, 5)

# Fall 2: Delta = (Eine (doppelte) Nullstelle)
a2, b2, c2 = 1, -2, 1 # (x-1)²
y2 = a2*x**2 + b2*x + c2
delta2 = b2**2 - 4*a2*c2
root2 = -b2/(2*a2)

axes[1].plot(x, y2, 'b-', linewidth=2)
axes[1].axhline(0, color='k', linewidth=0.5)
axes[1].axvline(0, color='k', linewidth=0.5)
axes[1].plot(root2, 0, 'ro', markersize=8)
axes[1].grid(True, alpha=0.3)
axes[1].set_title(f'$\Delta$ = {delta2} = 0\n(Eine doppelte Nullstellen)', fontsize=14)
axes[1].set_ylim(-0.5, 5)

# Fall 3: Delta < 0 (keine reellen Nullstellen)
a3, b3, c3 = 1, 0, 2
y3 = a3*x**2 + b3*x + c3
delta3 = b3**2 - 4*a3*c3

axes[2].plot(x, y3, 'b-', linewidth=2)
axes[2].axhline(0, color='k', linewidth=0.5)
axes[2].axvline(0, color='k', linewidth=0.5)
axes[2].grid(True, alpha=0.3)
axes[2].set_title(f'$\Delta$ = {delta3} < 0\n(Keien reellen Nullstellen)', fontsize=14)
axes[2].set_ylim(-0.5, 5)

for ax in axes:
    ax.set_xlabel('x')
    ax.set_ylabel('f(x)')

plt.tight_layout()
plt.show()

```

## Umwandlung zwischen den Formen

### Normalform → Scheitelpunktform (Quadratische Ergänzung)

Gegeben: $f(x) = ax^2 + bx + c$

**Schritte**:
1. Faktor $a$ ausklammern: $f(x) = a(x^2 + \frac{b}{a}x) + c$
2. Quadratische Ergänzung: $f(x) = a(x^2 + \frac{b}{a}x + (\frac{b}{2a})^2 - (\frac{b}{2a})^2) + c$
3. Vereinfachen: $f(x) = a(x + \frac{b}{2a})^2 - \frac{b^2}{4a} + c$
4. Umschreiben mit Scheitelpunkt $(h, k)$: $f(x) = a(x - h)^2 + k$

wobei $h = -\frac{b}{2a}$ und $k = c - \frac{b^2}{4a}$

### Normalform → Faktorisierte Form

1. Nullstellen finden mit der ABC-Formel
2. $f(x) = a(x - r_1)(x - r_2)$

### Scheitelpunktform → Normalform

Einfach ausmultiplizieren:
$$f(x) = a(x - h)^2 + k = a(x^2 - 2hx + h^2) + k = ax^2 - 2ahx + (ah^2 + k)$$

```python
import numpy as np
import sympy as sp

def convert_forms(a, b, c):
    """
    Konvertiert eine quadratische Funktion zwischen allen Formen.
    """
    print("=" * 60)
    print("QUADRATISCHE FUNKTIONSFORMEN - KONVERTER")
    print("=" * 60)

    # Normalform
    print(f"\n1. Normalform: f(x) = {a}x² + ({b})x + {c}")

    # Scheitelpunktform
    h = -b / (2*a)
    k = c - (b**2) / (4*a)
    print(f"\n2. Scheitelpunktform: f(x) = {a}(x - ({h}))² + ({k})")
    print(f"     Scheitelpunkt S({h}, {k})")

    # Nullstellen (für faktorisierte Form)
    discriminant = b**2 - 4*a*c
    print(f"\n3: Diskriminante: Δ = b² - 4ac = {discriminant}")

    if discriminant > 0:
        r1 = (-b - np.sqrt(discriminant)) / (2*a)
        r2 = (-b + np.sqrt(discriminant)) / (2*a)
        print(f"\n4. Fkatorisierte Form: f(x) = {a}(x - ({r1}))(x - ({r2})")
        print(f"    Nullstellen: x₁ = {r1}, x₂ = {r2}")
    elif discriminant == 0:
        r = -b / (2*a)
        print(f"\n4. Fkatorisierte Form: f(x) = {a}(x - ({r}))²")
        print(f"    Doppelte Nullstelle: x = {r}")
    else:
        print(f"\n4. Faktorisierte Form: Keine reellen Nullstellen (Δ < 0)")

    # Zusätzliche Informationen
    print(f"\n5. Weitere Eigenschaften:")
    print(f"   * y-Achsenabschnitt: (0, {c})")
    print(f"   * Symmetrieachse: x = {h}")
    print(f"   * Öffnungsrichtung: {'Nach oben' if a > 0 else 'Nach unten'} ")
    print(f"   • Scheitelpunkt ist {'Minimum' if a > 0 else 'Maximum'} ")
    print("=" * 60)

# Beispiele
print("\nBEISPIEL 1:")
convert_forms(1, -4, 3)

print("\nBEISPIEL 2:")
convert_forms(1, -2, 1)

print("\nBEISPIEL 3:")
convert_forms(-2, 8, -8)
```

## Anwendungen Quadratischer Funktionen

### 1. Physik: Projektilbewegung
Die Höhe eines geworfenen Objekts folgt einer quadratischen Funktion:
$$h(t) = h_0 + v_0 \,t - \frac{1}{2}g \,t^2$$

wobei:
- $h_0$ = Anfangshöhe
- $v_0$ = Anfangsgeschwindigkeit
- $g$ = Gravitationsbeschleunigung (=~ 9.81 m/s²)

Die Nullstellen zeigen, wann das Objekt den Boden berührt.

### 2. Geometrie: Flächenoptimierung
Bei rechteckigen Grundstücken mit begrenztem Umfang ist die Fläche eine quadratische Funktion:
$$A(x) = x(U/2 - x)$$

Der Scheitelpunkt zeigt die Abmessungen für maximale Fläche.

### 3. Optik & Akustik: Parabolische Reflektoren
Die Form von Satellitenschüsseln, Spiegeln und Mikrofonen folgt Parabeln.

```python
import numpy as np
import matplotlib.pyplot as plt

def solve_projectile_problem():
    """
    Ein Ball wird von einer Brücke (5m hoch) mit 10m/s geworfen.
    Wann kommt der Ball auf dem Boden an?
    """

    print("PROBLEM: Projektilbwewegung")
    print("-" * 50)

    h0 = 5    # Anfangshöhe in Metern
    v0 = 10   # Anfangsgeschwindigkeit in m/s
    g = 9.81  # Gravitation

    # h(t) = h0 + v0*t - 0.5*g*t²
    # h(t) = 5 + 10*t - 4.905*t²

    a = -g/2
    b = v0
    c = h0
    
    print(f"h(t) = {c} + {b}t + {a}t²")
    print(f"h(t) = {a}t² + {b}t + {c}")

    # Nullstellen finden
    discriminant = b**2 - 4*a*c
    t1 = (-b - np.sqrt(discriminant)) / (2*a)
    t2 = (-b + np.sqrt(discriminant)) / (2*a)

    print(f"\nDiskriminante: Δ = {discriminant:.2f}")
    print(f"Nullstellen: t1 = {t1:.2f}, t2 = {t2:.2f}s")
    print(f"\n Der Ball kommt nach {max(t1, t2):.2f} Sekunden auf dem Boden an.") 

    t = np.linspace(0, max(t1, t2) + 0.5, 100)
    h = c + b*t + a*t**2

    plt.figure(figsize=(10,6))
    plt.plot(t, h, 'b-', linewidth=2, label='h(t) = 5 + 10t - 4.905t²')
    plt.axhline(0, color='k', linewidth=0.5)
    plt.plot(max(t1, t2), 0, 'ro', markersize=10, label=f'Aufprall: t = {max(t1, t2):.2f}s')

    # Maximale Höhe (Scheitelpunkt)
    t_max = -b / (2*a)
    h_max = c + b*t_max + a*t_max**2
    plt.plot(t_max, h_max, 'go', markersize=10, label=f'Max. Höhe: h = {h_max:.2f}m bei t = {t_max:.2f}s')

    plt.xlabel('Zeit (s)')
    plt.ylabel('Höhe (m)')
    plt.title('Projektilbewegung: Ball von Brücke')
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

def solve_optimization_problem():
    """
    Ein Bauer hat 40m Zaun für ein reckteckiges Gehege.
    Wie sind die optimalen Abmessungen?
    """
    print("\n\nPROBLEM: Flächenoptimierung")
    print("-" * 50)

    perimeter = 40 # Gesamtzaun

    # Wenn Länge = x, dann Breite = (P/2 - x) = (20 - x)
    # Fläche A(x) = x * (20 - x) = -x² + 20x

    a = -1
    b = 20
    c = 0

    print(f"Umnfang = {perimeter}m")
    print(f"A(x) = x(20 . x ) = {a}x² + {b}x + {c}")

    # Scheitelpunkt (Maximum)
    x_optimal = -b / (2*a)
    area_max = c + b*x_optimal + a*x_optimal**2

    print(f"\nOptimale Länge: x = {x_optimal}m")
    print(f"Optimale Breite: {20 - x_optimal}m")
    print(f"Maximale Fläche: A = {area_max}m²")

    # Visualisierung
    x = np.linspace(0, 20, 100)
    area = -x**2 + 20*x

    plt.figure(figsize=(10,6))
    plt.plot(x, area, 'b-', linewidth=2, label='A(x) = -x² + 20x')
    plt.plot(x_optimal, area_max, 'ro', markersize=10, label=f'Maximum: A = {area_max}m² bei x = {x_optimal}m')
    plt.axvline(x_optimal, color='r', linestyle='--', alpha=0.5)

    plt.xlabel('Länge (m)')
    plt.ylabel('Fläche (m²)')
    plt.title('Flächenoptimierung: Rechteckiges Gehege mit 40m Zaun')
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

solve_projectile_problem()
solve_optimization_problem()
```

## Ausblick: Verbindung zur Analysis

### Ableitungen (Differentialrechnung)
Die Ableitung einer quadratischen Funktion ist linear:
$$f(x) = ax^2 + bx + c$$
$$f'(x) = 2ax + b$$

Der Scheitelpunkt liegt dort, wo $f'(x) = 0$:
$$2ax + b = 0 \Rightarrow x = -\frac{b}{2a}$$

Das ist genau die Formel, die wir bereits kennen!

### Integrale (Integralrechnung)
Die Fläche unter einer Parabel berechnet sich durch Integration:
$$\int_a^b (ax^2 + bx + c) dx = \left[\frac{a}{3}x^3 + \frac{b}{2}x^2 + cx\right]_a^b$$

### Taylor-Reihen
Jede glatte Funktion kann in ihrer Näherung durch eine quadratische Funktion (quadratische Taylor-Näherung) angenähert werden.

## Zusammenfassung & Tipps

### Prüfungscheckliste

- [ ] **Definition**: $f(x) = ax^2 + bx + c$ mit $a \neq 0$
- [ ] **Drei Formen**: Normalform, Scheitelpunktform, faktorisierte Form
- [ ] **Scheitelpunkt**: $h = -\frac{b}{2a}$, $k = c - \frac{b^2}{4a}$
- [ ] **Diskriminante**: $\Delta = b^2 - 4ac$
- [ ] **ABC-Formel**: $x = \frac{-b \pm \sqrt{\Delta}}{2a}$
- [ ] **Quadratische Ergänzung**: Umwandlung zwischen Formen
- [ ] **Eigenschaften**: Öffnungsrichtung, Symmetrie, Extremwerte

### Wichtige Formeln (zum Auswendiglernen)

| Formel | Verwendung |
|--------|-----------|
| $x_s = -\frac{b}{2a}$ | x-Koordinate des Scheitelpunkts |
| $y_s = c - \frac{b^2}{4a}$ | y-Koordinate des Scheitelpunkts |
| $\Delta = b^2 - 4ac$ | Diskriminante |
| $x = \frac{-b \pm \sqrt{\Delta}}{2a}$ | ABC-Formel (Nullstellen) |

### Schnelle Checks

1. **Ist es quadratisch?** → Prüfe, ob $a \neq 0$
2. **Wie viele Nullstellen?** → Berechne $\Delta$
3. **Wo ist der Extremwert?** → Benutze $x_s = -\frac{b}{2a}$
4. **Wie breit ist die Parabel?** → Schau auf $|a|$
5. **Öffnet sie oben oder unten?** → Schau auf das Vorzeichen von $a$
