# Analysis
## 03 - Ableitungsregeln

Effizientes Ableiten ohne Grenzwert:
- Potenzregel ($(x^n)' = nx^{n-1}$)
- Konstanten-/Faktor-/Summenregel, Produktregel ($(uv)' = u'v + uv'$)
- Quotientenregel und Kettenregel ($(f(g(x)))' = f'(g(x)) \cdot g'(x)$) für zusammengesetzte Funktionen.

> Vorausgesetzt: Ableitungsbegriff.

### Lernziele

- **Die Potenzregel** — die wichtigste Grundregel
- **Konstanten- und Faktorregel** — Ableitungen von Konstanten und Vielfachen
- **Summen- und Differenzregel** — Term für Term ableiten
- **Die Produktregel** — Ableitung von Produkten zweier Funktionen
- **Die Quotientenregel** — Ableitung von Brüchen
- **Die Kettenregel** — Ableitung verschachtelter (zusammengesetzter) Funktionen

---

### 1. Die Potenzregel

Die **Potenzregel** ist die am häufigsten verwendete Ableitungsregel. Sie gilt für Funktionen der Form $f(x) = x^n$:

$$\boxed{f(x) = x^n \quad \Rightarrow \quad f'(x) = n \cdot x^{n-1}}$$

In Worten: **Der Exponent wird als Faktor vorangestellt und anschließend um eins vermindert.**

Beispiele:
- $f(x) = x^2 \Rightarrow f'(x) = 2x^1 = 2x$
- $f(x) = x^3 \Rightarrow f'(x) = 3x^2$
- $f(x) = x^1 = x \Rightarrow f'(x) = 1 \cdot x^0 = 1$ (Steigung der Geraden $y=x$ ist immer 1)
- $f(x) = x^{-1} = \frac{1}{x} \Rightarrow f'(x) = -1 \cdot x^{-2} = -\frac{1}{x^2}$
- $f(x) = \sqrt{x} = x^{1/2} \Rightarrow f'(x) = \frac{1}{2} x^{-1/2} = \frac{1}{2\sqrt{x}}$

### 2. Konstanten- und Faktorregel

**Ableitung einer Konstanten:** Wenn $f(x) = c$ (eine horizontale Linie), dann ist die Steigung überall null:
$$\boxed{f(x) = c \quad \Rightarrow \quad f'(x) = 0}$$

**Konstanter Faktor:** Ein konstanter Faktor bleibt beim Ableiten erhalten:
$$\boxed{f(x) = c \cdot g(x) \quad \Rightarrow \quad f'(x) = c \cdot g'(x)}$$

Beispiele:
- $f(x) = 5 \Rightarrow f'(x) = 0$
- $f(x) = 7x^2 \Rightarrow f'(x) = 7 \cdot 2x = 14x$
- $f(x) = -3x^4 \Rightarrow f'(x) = -3 \cdot 4x^3 = -12x^3$

### 3. Summen- und Differenzregel

Man kann **Term für Term** ableiten:
$$\boxed{[f(x) \pm g(x)]' = f'(x) \pm g'(x)}$$

Damit kannst Du jedes Polynom ableiten!

**Beispiel:** $h(x) = x^3 + 2x - 5$
- Ableitung von $x^3$ ist $3x^2$
- Ableitung von $2x$ ist $2$
- Ableitung von $-5$ ist $0$
- Also: $h'(x) = 3x^2 + 2$

```python
import sympy as sp

x = sp.Symbol('x')

# --- Potenzregel, Faktorregel, Summenregel mit SymPy verifizieren ---

regeln_beispiele = {
    "Potenzregel: x⁴":          x**4,
    "Potenzregel: √x":          sp.sqrt(x),
    "Potenzregel: 1/x²":        x**(-2),
    "Faktorregel: 7x²":         7*x**2,
    "Summenregel: x³ + 2x - 5": x**3 + 2*x - 5,
    "Polynom: 4x⁵ - 3x² + x":  4*x**5 - 3*x**2 + x,
}

print("Überprüfung der Grundregeln mit SymPy:")
print("=" * 50)
for name, funktion in regeln_beispiele.items():
    ableitung = sp.diff(funktion, x)
    print(f"{name}")
    print(f"  f(x)  = {funktion}")
    print(f"  f'(x) = {ableitung}")
    print()
```

### 4. Die Produktregel

Wenn $f(x) = u(x) \cdot v(x)$, also ein Produkt zweier Funktionen, dann gilt:

$$\boxed{[u(x) \cdot v(x)]' = u'(x) \cdot v(x) + u(x) \cdot v'(x)}$$

*Merkhilfe: „Erste abgeleitet mal Zweite plus Erste mal Zweite abgeleitet."*

**Beispiel:** $f(x) = x^2 \cdot \sin(x)$
- $u(x) = x^2 \quad \Rightarrow \quad u'(x) = 2x$
- $v(x) = \sin(x) \quad \Rightarrow \quad v'(x) = \cos(x)$
- $f'(x) = 2x \cdot \sin(x) + x^2 \cdot \cos(x)$

---

### 5. Die Quotientenregel

Wenn $f(x) = \frac{u(x)}{v(x)}$, ein Bruch zweier Funktionen, dann gilt:

$$\boxed{\left[\frac{u(x)}{v(x)}\right]' = \frac{u'(x) \cdot v(x) - u(x) \cdot v'(x)}{[v(x)]^2}}$$

*Merkhilfe: „Zähler abgeleitet mal Nenner minus Zähler mal Nenner abgeleitet - alles durch Nenner zum Quadrat."* (NAZ - NZA durch N²)

**Beispiel:** $f(x) = \frac{x^2 + 1}{x - 3}$
- $u(x) = x^2 + 1 \quad \Rightarrow \quad u'(x) = 2x$
- $v(x) = x - 3 \quad \Rightarrow \quad v'(x) = 1$
- $f'(x) = \frac{2x(x-3) - (x^2+1) \cdot 1}{(x-3)^2} = \frac{2x^2 - 6x - x^2 - 1}{(x-3)^2} = \frac{x^2 - 6x - 1}{(x-3)^2}$

```python
import sympy as sp

x = sp.Symbol('x')

# --- Produktregel verifizieren ---
print("=== Produktregel ===")
u1 = x**2
v1 = sp.sin(x)
produkt = u1 * v1
produkt_abl = sp.diff(produkt, x)
print(f"f(x)  = {produkt}")
print(f"f'(x) = {produkt_abl}")
print(f"Vereinfacht: {sp.simplify(produkt_abl)}")
print()

# --- Quotientenregel verifizieren ---
print("=== Quotientenregel ===")
u2 = x**2 + 1
v2 = x - 3
quotient = u2 / v2
quotient_abl = sp.diff(quotient, x)
print(f"f(x)  = {quotient}")
print(f"f'(x) = {quotient_abl}")
print(f"Vereinfacht: {sp.simplify(quotient_abl)}")
print()

# Wert der Ableitung an einer Stelle
print(f"f'(5) = {quotient_abl.subs(x, 5)}")
```

### 6. Die Kettenregel

Die Kettenregel brauchst Du, wenn eine Funktion **verschachtelt** (zusammengesetzt) ist, z.B. $f(x) = (x^2 + 1)^3$ oder $h(x) = \sin(x^2)$.

Sei $y = f(g(x))$ mit:
- $f$ = **äußere Funktion**
- $g$ = **innere Funktion**

Dann gilt die Kettenregel:

$$\boxed{[f(g(x))]' = f'(g(x)) \cdot g'(x)}$$

In Worten: **„Äußere Funktion ableiten (innere stehen lassen), dann mal die Ableitung der inneren Funktion."**

In Leibniz-Notation besonders anschaulich:
$$\frac{dy}{dx} = \frac{dy}{du} \cdot \frac{du}{dx}$$

#### Beispiel 1: $y = (x^2 + 1)^3$
1. **Funktionen identifizieren:**
   - Äußere Funktion: $f(u) = u^3$ (wobei $u$ der „innere Teil" ist)
   - Innere Funktion: $g(x) = x^2 + 1$
2. **Ableitungen bilden:**
   - $f'(u) = 3u^2$ (Potenzregel)
   - $g'(x) = 2x$ (Potenzregel)
3. **Kettenregel anwenden:** $f'(g(x)) \cdot g'(x)$
   - $f'(g(x)) = 3(x^2+1)^2$ (innere Funktion für $u$ einsetzen)
   - Mal $g'(x) = 2x$
4. **Ergebnis:**
   - $y' = 3(x^2+1)^2 \cdot 2x = 6x(x^2+1)^2$

#### Beispiel 2: $h(x) = \sin(x^2)$
- Äußere: $\sin(u) \Rightarrow \cos(u)$
- Innere: $x^2 \Rightarrow 2x$
- Ergebnis: $h'(x) = \cos(x^2) \cdot 2x = 2x\cos(x^2)$

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.Symbol('x')

# --- Kettenregel verifizieren ---
print("=== Kettenregel: Beispiel 1 ===")
h1_sym = (x**2 + 1)**3
h1_abl = sp.diff(h1_sym, x)
print(f"h(x)  = {h1_sym}")
print(f"h'(x) = {h1_abl}")
print(f"Faktorisiert: {sp.factor(h1_abl)}")
print()

print("=== Kettenregel: Beispiel 2 ===")
h2_sym = sp.sin(x**2)
h2_abl = sp.diff(h2_sym, x)
print(f"h(x)  = {h2_sym}")
print(f"h'(x) = {h2_abl}")
print()

# --- Visuelle Intuition der Kettenregel ---
# Für h(x) = sin(x²): Wie verändert sich die Steigung durch die Verkettung?

x_sym, u_sym = sp.symbols('x u')

# Innere Funktion g(x) = x²
g_sym = x_sym**2
g_abl_sym = sp.diff(g_sym, x_sym)

# Äußere Funktion f(u) = sin(u)
f_sym = sp.sin(u_sym)
f_abl_sym = sp.diff(f_sym, u_sym)

# Zusammengesetzte Funktion h(x) = sin(x²)
h_sym = sp.sin(x_sym**2)
h_abl_sym = sp.diff(h_sym, x_sym)

# In numerische Funktionen umwandeln
g = sp.lambdify(x_sym, g_sym, 'numpy')
g_abl = sp.lambdify(x_sym, g_abl_sym, 'numpy')
f = sp.lambdify(u_sym, f_sym, 'numpy')
f_abl = sp.lambdify(u_sym, f_abl_sym, 'numpy')
h = sp.lambdify(x_sym, h_sym, 'numpy')
h_abl = sp.lambdify(x_sym, h_abl_sym, 'numpy')

# --- Punkt für die Visualisierung ---
x_punkt = 1.25
u_punkt = g(x_punkt)

# --- Plot: Drei Panels ---
fig = plt.figure(figsize=(22, 14))
fig.suptitle(f"Kettenregel-Visualisierung bei x = {x_punkt}", fontsize=20)

# Panel 1: Innere Funktion g(x) = x²
ax1 = plt.subplot(2, 2, 1)
x_vals = np.linspace(0, 10, 100)
ax1.plot(x_vals, g(x_vals), label=r"$g(x) = x^2$ (innere)", color='blue')
ax1.set_title(r"1. Innere Funktion: $g'(x)$", fontsize=14)
ax1.set_xlabel('x')
ax1.set_ylabel('u')
ax1.grid(True)

steigung_g = g_abl(x_punkt)
tangente_g = steigung_g * (x_vals - x_punkt) + u_punkt
ax1.plot(x_vals, tangente_g, 'r--', label=f"$g'({x_punkt}) = {steigung_g:.2f}$")
ax1.scatter(x_punkt, u_punkt, color='red', s=80, zorder=5)
ax1.legend(fontsize=11)

# Panel 2: Äußere Funktion f(u) = sin(u)
ax2 = plt.subplot(2, 2, 2)
u_vals = np.linspace(0, 10, 100)
ax2.plot(u_vals, f(u_vals), label=r"$f(u) = \sin(u)$ (äußere)", color='green')
ax2.set_title(r"2. Äußere Funktion: $f'(g(x))$", fontsize=14)
ax2.set_xlabel('u')
ax2.set_ylabel('y')
ax2.grid(True)

steigung_f = f_abl(u_punkt)
tangente_f = steigung_f * (u_vals - u_punkt) + f(u_punkt)
ax2.plot(u_vals, tangente_f, 'm--', label=f"$f'({u_punkt:.2f}) = {steigung_f:.2f}$")
ax2.scatter(u_punkt, f(u_punkt), color='magenta', s=80, zorder=5)
ax2.legend(fontsize=11)

# Panel 3: Zusammengesetzte Funktion h(x) = sin(x²)
ax3 = plt.subplot(2, 1, 2)
x_vals_h = np.linspace(0, 4, 400)
ax3.plot(x_vals_h, h(x_vals_h), label=r"$h(x) = \sin(x^2)$", color='purple')
ax3.set_title(r"3. Verkettete Funktion: $h'(x) = f'(g(x)) \cdot g'(x)$", fontsize=14)
ax3.set_xlabel('x')
ax3.set_ylabel('y')
ax3.grid(True)

steigung_h = h_abl(x_punkt)
tangente_h = steigung_h * (x_vals_h - x_punkt) + h(x_punkt)
ax3.plot(x_vals_h, tangente_h, 'k--', label=f"$h'({x_punkt}) = {steigung_h:.2f}$")
ax3.scatter(x_punkt, h(x_punkt), color='black', s=80, zorder=5)
ax3.legend(fontsize=11)

# Verifikationstext
produkt_steigungen = steigung_g * steigung_f
ax3.text(0.5, 0.1,
         f"Produkt der Steigungen: $g' \\cdot f' = ({steigung_g:.2f}) \\cdot ({steigung_f:.2f}) = {produkt_steigungen:.2f}$",
         transform=ax3.transAxes, fontsize=13, bbox=dict(facecolor='yellow', alpha=0.5))

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

### 7. Übersicht: Alle Regeln auf einen Blick

| Regel | Formel | Beispiel |
|-------|--------|----------|
| **Potenzregel** | $(x^n)' = n \cdot x^{n-1}$ | $(x^5)' = 5x^4$ |
| **Konstantenregel** | $(c)' = 0$ | $(7)' = 0$ |
| **Faktorregel** | $(c \cdot f)' = c \cdot f'$ | $(3x^2)' = 6x$ |
| **Summenregel** | $(f \pm g)' = f' \pm g'$ | $(x^3 + x)' = 3x^2 + 1$ |
| **Produktregel** | $(u \cdot v)' = u'v + uv'$ | $(x^2 \sin x)' = 2x\sin x + x^2\cos x$ |
| **Quotientenregel** | $\left(\frac{u}{v}\right)' = \frac{u'v - uv'}{v^2}$ | $\left(\frac{x}{x+1}\right)' = \frac{1}{(x+1)^2}$ |
| **Kettenregel** | $[f(g(x))]' = f'(g(x)) \cdot g'(x)$ | $[\sin(x^2)]' = 2x\cos(x^2)$ |

#### Wichtige Standardableitungen

Einige Ableitungen solltest Du auswendig kennen:

| Funktion $f(x)$ | Ableitung $f'(x)$ |
|---|---|
| $x^n$ | $nx^{n-1}$ |
| $e^x$ | $e^x$ |
| $\ln(x)$ | $\frac{1}{x}$ |
| $\sin(x)$ | $\cos(x)$ |
| $\cos(x)$ | $-\sin(x)$ |
| $\tan(x)$ | $\frac{1}{\cos^2(x)}$ |
| $a^x$ | $a^x \cdot \ln(a)$ |
