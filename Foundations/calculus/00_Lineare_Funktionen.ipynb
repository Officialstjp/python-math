Bei einer lineraren Steigung ist der Anstieg über die horizontale Entfernung konstant.



$$
k = \frac{\Delta y}{\Delta x} = \frac{f(x_2) - f(x_1)}{x_2 - x_1} \quad \text{konstant}
$$


Die Steigung wird auch als 

Differenzenquotient, mittlere Änderungsrate oder als Steigung mithilfe des Steigungsdreiecks bezeichnet.
