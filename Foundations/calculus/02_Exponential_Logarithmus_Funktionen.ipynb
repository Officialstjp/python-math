### Euler Identität
Die "Euler Identität" stellt einen einfachen Zusammenhang zwischen vier bedeutendsten mathematischen Konstanten her her:
$$
e^{i \cdot \pi} + 1 = 0
$$

wobei:
- $e$: Eulersche Zahl
- $i$: Imaginäre Einheit i der komplexen Zahlen
- $\pi$: Kreiszahl
- $0$ und $1$
