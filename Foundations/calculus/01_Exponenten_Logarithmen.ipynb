## Logarithmen

**Definition:**

Der Logarithmus ist die Umkehrfunktion zur Exponentialfunktion. Er wird angeschrieben als:

$$...
$$

gesprochen: Der Logarithmus zur Basis ...

Der Logarithmus aus der Basis der eulerschen Zahl wird der "**natürliche Logarithmus**" (Logarithmus Naturales) genannt.

Angeschrieben als:
$$
\log_e = \ln
$$

Früher wurde auch oft der "Logarithmus Duales" in Geräten gesehen (angeschrieben: $ld$). Heute wird dieser Aufgrund von Verbesserungen in Rechenleistung kaum noch benötigt.
