## Logarithmen

**Definition:**

Der Logarithmus ist die Umkehrfunktion zur Exponentialfunktion. Er wird angeschrieben als:

$$...
$$

gesprochen: Der Logarithmus zur Basis ...

Der Logarithmus aus der Basis der eulerschen Zahl wird der "**natürliche Logarithmus**" (Logarithmus Naturales) genannt.

Angeschrieben als:
$$
\log_e = \ln
$$

Früher wurde auch oft der "Logarithmus Duales" in Geräten gesehen (angeschrieben: $ld$). Heute wird dieser Aufgrund von Verbesserungen in Rechenleistung kaum noch benötigt.



### Rechenregeln für Logarithmen

Die wichtigsten Rechenregeln für Logarithmen sind:

$$
\log (a^x) = x \cdot \log(a)
$$

$$
\log(a \cdot b) = \log(a) + \log(b)
$$

$$
\log\left(\frac{a}{b}\right) = \log(a) - \log(b)
$$


Beispiel:

$$
\log(10^3) = 3 \cdot \log(10) = 3 \cdot 1 = 3
$$

$$
\log_2(64) = \log_2(2^6) = 6 \cdot \log_2(2) = 6 \cdot 1 = 6
$$

$$
\log(1000) = \log(10^3) = 3 \cdot \log(10) = 3 \cdot 1 = 3
$$

$$
\log\left(\frac{100}{10}\right) = \log(100) - \log(10) = 2 - 1 = 1
$$

## Wechsel der Logarithmus-Basis

Der Logarithmus zur Basis $a$ kann in einen Logarithmus zur Basis $b$ umgerechnet werden:
$$
\log_a(x) = \frac{\log_b(x)}{\log_b(a)}
$$

Ein klassisches Beispiel ist die Umrechnung in den natürlichen Logarithmus:
$$
\log_a(x) = \frac{\ln(x)}{\ln(a)}
$$
