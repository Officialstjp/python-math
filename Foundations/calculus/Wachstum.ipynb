## Berechnung einer Wachstumsperiode

Wollen wir nun wissen nach wie vielen Zeiteinheiten ein gewisser Endwert $N(t)$ erreicht wird, so können wir die Formel umstellen:

### Herleitung

Mit der allgemeinen Wachstumsformel: $N(t) = N_0 \cdot a^t$

Dividieren wir beide Seiten durch $N_0$:
$$
\frac{N(t)}{N_0} = a^t
$$

Und ziehen nun den Logarithmus zur Basis $a$ auf beiden Seiten:
$$
\log_a\left(\frac{N(t)}{N_0}\right) = \log_a\left(a^t\right)
$$

Das vereinfacht sich zu:
$$
\log_a\left(\frac{N(t)}{N_0}\right) = t
$$

### Formel

Nun können wir sagen, um eine Wachstumsperiode zu berechnen, nimmt man den Logarithmus zur Basis des Wachstumsfaktors $a$ aus dem Verhältnis der Endanzahl $N(t)$ zur Anfangsanzahl $N_0$:

$$
\log_a\left(\frac{N(t)}{N_0}\right) = T 
$$

---


### Änderungsrate

Die **relative Änderungsrate** (Wachstumsrate) $r$ kann berechnet werden mit:
$$
r = \frac{N(t) - N_0}{N_0 \cdot t}
$$
wobei diese dem Verhätlnis des Wachstumsfaktors minus 1 entspricht:
$$
r = \frac{a - 1}{t}
$$
